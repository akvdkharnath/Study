LRUCache

````python
from collections import OrderedDict

class LRUCache:
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.cache = OrderedDict()
    
    def get(self, key):
        if key not in self.cache:
            return -1
        
        # Move to end (most recently used)
        self.cache.move_to_end(key)
        return self.cache[key]
    
    def put(self, key, value):
        if key in self.cache:
            # Update value and move to end
            self.cache.move_to_end(key)
        else:
            # Check capacity
            if len(self.cache) >= self.capacity:
                # Evict least recently used (first item)
                self.cache.popitem(last=False)
        
        self.cache[key] = value

# Test it
cache = LRUCache(2)

cache.put(1, 1)  # cache = {1:1}
cache.put(2, 2)  # cache = {2:2, 1:1}
print(cache.get(1))  # 1 (moves 1 to end: {2:2, 1:1})
cache.put(3, 3)  # Evicts 2: {1:1, 3:3}
print(cache.get(2))  # -1 (not found)
print(cache.get(1))  # 1
print(cache.get(3))  # 3
````

## Problem 1: Two Sum

**Problem Statement**: Given an array of integers and a target sum, find two numbers that add up to the target. Return their indices.

**Example**:
- Input: nums = [2, 7, 11, 15], target = 9
- Output: [0, 1]
- Explanation: nums[0] + nums[1] = 2 + 7 = 9

**Constraints**:
- You may assume that each input would have exactly one solution
- You may not use the same element twice
- Return the answer in any order


### Solution 1: Brute Force Approach (Simple)

**Time Complexity**: O(n²)  
**Space Complexity**: O(1)

**Approach**: 
- Use nested loops to check every pair of numbers
- For each element, check if there's another element that sums to target
- Return indices when found


````python
def two_sum_brute_force(nums, target):
    """
    Brute Force Solution - O(n²) time, O(1) space
    Simple but inefficient approach
    """
    n = len(nums)
    
    for i in range(n):
        for j in range(i + 1, n):
            if nums[i] + nums[j] == target:
                return [i, j]
    
    return []  # No solution found

# Test the brute force solution
test_cases = [
    ([2, 7, 11, 15], 9),
    ([3, 2, 4], 6),
    ([3, 3], 6),
    ([1, 2, 3, 4, 5], 8)
]

print("Brute Force Solution Results:")
for nums, target in test_cases:
    result = two_sum_brute_force(nums, target)
    print(f"nums = {nums}, target = {target} -> {result}")
    if result:
        print(f"  Verification: {nums[result[0]]} + {nums[result[1]]} = {nums[result[0]] + nums[result[1]]}")
    print()
````

### Solution 2: Hash Map Approach (Most Effective)

**Time Complexity**: O(n)  
**Space Complexity**: O(n)

**Approach**: 
- Use a hash map to store numbers and their indices
- For each number, check if (target - current_number) exists in the map
- If found, return the indices; otherwise, add current number to map
- This is the most efficient solution for this problem


````python
#    ([2, 7, 11, 15], 9),

# test_cases = [
#     ([2, 7, 11, 15], 9),
#     ([3, 2, 4], 6),
#     ([3, 3], 6),
#     ([1, 2, 3, 4, 5], 8)
# ]


def two_sum_hash_map(nums, target):
    """
    Hash Map Solution - O(n) time, O(n) space
    Most efficient approach using dictionary
    """
    num_to_index = {}
    
    for i, num in enumerate(nums):
        complement = target - num
        
        # Check if complement exists in our map
        if complement in num_to_index:
            return [num_to_index[complement], i]
        
        # Store current number and its index
        num_to_index[num] = i
    
    return []  # No solution found

# Test the hash map solution
print("Hash Map Solution Results:")
for nums, target in test_cases:
    result = two_sum_hash_map(nums, target)
    print(f"nums = {nums}, target = {target} -> {result}")
    if result:
        print(f"  Verification: {nums[result[0]]} + {nums[result[1]]} = {nums[result[0]] + nums[result[1]]}")
    print()
````

## Problem 2: Maximum Subarray Sum (Kadane's Algorithm)

**Problem Statement**: Find the contiguous subarray within an array that has the largest sum. Return the maximum sum.

**Example**:
- Input: nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
- Output: 6
- Explanation: The subarray [4, -1, 2, 1] has the largest sum = 6

**Constraints**:
- 1 ≤ nums.length ≤ 10⁵
- -10⁴ ≤ nums[i] ≤ 10⁴
- The array contains at least one number

**Follow-up**: Can you solve it in O(1) space complexity?


### Solution 1: Brute Force Approach (All Possible Subarrays)

**Time Complexity**: O(n³)  
**Space Complexity**: O(1)

**Approach**: 
- Generate all possible contiguous subarrays
- Calculate sum for each subarray
- Keep track of the maximum sum found
- This is the most straightforward but inefficient approach


````python
def max_subarray_brute_force(nums):
    """
    Brute Force Solution - O(n³) time, O(1) space
    Check all possible contiguous subarrays
    """
    n = len(nums)
    max_sum = float('-inf')
    
    # Generate all possible subarrays
    for i in range(n):           # Start index
        for j in range(i, n):    # End index
            # Calculate sum of subarray from i to j
            current_sum = 0
            for k in range(i, j + 1):
                current_sum += nums[k]
            
            # Update maximum sum
            max_sum = max(max_sum, current_sum)
    
    return max_sum

# Test the brute force solution
test_cases = [
    [-2, 1, -3, 4, -1, 2, 1, -5, 4],  # Expected: 6
    [1],                               # Expected: 1
    [5, 4, -1, 7, 8],                 # Expected: 23
    [-1],                             # Expected: -1
    [-2, -1, -3],                     # Expected: -1
    [1, 2, 3, 4, 5]                   # Expected: 15
]

print("Brute Force Solution Results:")
for i, nums in enumerate(test_cases):
    result = max_subarray_brute_force(nums)
    print(f"Test {i+1}: nums = {nums}")
    print(f"         Max subarray sum = {result}")
    print()
````

### Solution 2: Kadane's Algorithm (Optimal)

**Time Complexity**: O(n)  
**Space Complexity**: O(1)

**Approach**: 
- Use dynamic programming concept
- At each position, decide whether to extend the existing subarray or start a new one
- Keep track of maximum sum ending at current position
- Update global maximum whenever we find a better sum
- This is the most efficient solution


````python
def max_subarray_kadane(nums):
    """
    Kadane's Algorithm - O(n) time, O(1) space
    Optimal solution using dynamic programming
    """
    n = len(nums)
    
    # Initialize variables
    max_ending_here = nums[0]  # Maximum sum ending at current position
    max_so_far = nums[0]       # Global maximum sum found so far
    
    # Process each element
    for i in range(1, n):
        # Decide: extend existing subarray or start new one
        max_ending_here = max(nums[i], max_ending_here + nums[i])
        
        # Update global maximum
        max_so_far = max(max_so_far, max_ending_here)
    
    return max_so_far

# Test Kadane's algorithm
print("Kadane's Algorithm Results:")
for i, nums in enumerate(test_cases):
    result = max_subarray_kadane(nums)
    print(f"Test {i+1}: nums = {nums}")
    print(f"         Max subarray sum = {result}")
    print()
````

````python
# Verify both solutions give the same results
print("Verification - Both solutions should give identical results:")
print("=" * 60)

for i, nums in enumerate(test_cases):
    brute_result = max_subarray_brute_force(nums)
    kadane_result = max_subarray_kadane(nums)
    
    print(f"Test {i+1}: nums = {nums}")
    print(f"  Brute Force: {brute_result}")
    print(f"  Kadane's:    {kadane_result}")
    print(f"  Match: {'✓' if brute_result == kadane_result else '✗'}")
    print()
````

### Key Insights:

   - Checks all possible contiguous subarrays

### Algorithm Intuition:

**Kadane's Algorithm Core Idea**:
- At each position, we have two choices:
  1. **Extend** the existing subarray (add current element)
  2. **Start fresh** with current element as new subarray
- We choose the option that gives us the maximum sum
- This ensures we always have the best subarray ending at current position

**Detailed Explanation:**

**Line 1: `n = len(nums)`**

**Line 2: `max_sum = float('-inf')`**

**Line 3: `for i in range(n):`**
- Outer loop that controls the **start index** of subarrays

- Adds each element of the current subarray to the sum
- This calculates the sum of subarray from index `i` to `j`

**Line 8: `max_sum = max(max_sum, current_sum)`**
- Updates the global maximum if current subarray sum is larger
- Uses `max()` function to choose the larger value

**Line 9: `return max_sum`**
- Returns the maximum sum found across all subarrays

### Why Three Nested Loops?

1. **First Loop (i)**: Controls where subarrays **start**
2. **Second Loop (j)**: Controls where subarrays **end** 
3. **Third Loop (k)**: Calculates the **sum** of current subarray

### Time Complexity Analysis

- **Outer loop**: runs `n` times
- **Middle loop**: runs `n-i` times for each `i` (average `n/2` times)
- **Inner loop**: runs `j-i+1` times for each `j` (average `n/2` times)

**Total**: `n × (n/2) × (n/2) = O(n³)`

### Why This Approach Works

The algorithm is **exhaustive** - it checks every possible contiguous subarray:
- Single elements: [1], [-2], [3]
- Pairs: [1, -2], [-2, 3] 
- Triplets: [1, -2, 3]

Since we check all possibilities, we're guaranteed to find the maximum sum!


### Enhanced Kadane's Algorithm - Return the Actual Subarray

**Question**: What if we need to return not just the maximum sum, but also the actual subarray that produces this sum?

**Answer**: Yes! We can modify Kadane's algorithm to track the start and end indices of the optimal subarray.


````python
def max_subarray_kadane_with_indices(nums):
    """
    Enhanced Kadane's Algorithm - Returns both sum and subarray indices
    Time Complexity: O(n), Space Complexity: O(1)
    """
    n = len(nums)
    
    # Initialize variables
    max_ending_here = nums[0]  # Maximum sum ending at current position
    max_so_far = nums[0]       # Global maximum sum found so far
    
    # Track indices for the optimal subarray
    start = 0                  # Start index of current best subarray
    end = 0                    # End index of current best subarray
    temp_start = 0             # Temporary start for current subarray
    
    # Process each element
    for i in range(1, n):
        # Decide: extend existing subarray or start new one
        if max_ending_here + nums[i] > nums[i]:
            # Extend existing subarray
            max_ending_here = max_ending_here + nums[i]
        else:
            # Start new subarray from current position
            max_ending_here = nums[i]
            temp_start = i
        
        # Update global maximum and indices
        if max_ending_here > max_so_far:
            max_so_far = max_ending_here
            start = temp_start
            end = i
    
    return max_so_far, start, end

def get_subarray_elements(nums, start, end):
    """Helper function to extract subarray elements"""
    return nums[start:end+1]

# Test the enhanced algorithm
print("Enhanced Kadane's Algorithm - With Subarray Indices:")
print("=" * 60)

test_cases = [
    [-2, 1, -3, 4, -1, 2, 1, -5, 4],  # Expected: 6, subarray [4, -1, 2, 1]
    [1],                               # Expected: 1, subarray [1]
    [5, 4, -1, 7, 8],                 # Expected: 23, subarray [5, 4, -1, 7, 8]
    [-1],                             # Expected: -1, subarray [-1]
    [-2, -1, -3],                     # Expected: -1, subarray [-1]
    [1, 2, 3, 4, 5]                   # Expected: 15, subarray [1, 2, 3, 4, 5]
]

for i, nums in enumerate(test_cases):
    max_sum, start_idx, end_idx = max_subarray_kadane_with_indices(nums)
    subarray = get_subarray_elements(nums, start_idx, end_idx)
    
    print(f"Test {i+1}: nums = {nums}")
    print(f"         Max sum = {max_sum}")
    print(f"         Subarray indices = [{start_idx}:{end_idx}]")
    print(f"         Subarray elements = {subarray}")
    print(f"         Verification: sum({subarray}) = {sum(subarray)}")
    print()
````

### Step-by-Step Explanation of Enhanced Algorithm

**Key Changes Made:**

1. **Added Index Tracking Variables:**
   - `start`: Start index of the globally best subarray
   - `end`: End index of the globally best subarray  
   - `temp_start`: Start index of current subarray being considered

2. **Modified Decision Logic:**
   - When extending: Keep `temp_start` unchanged
   - When starting fresh: Update `temp_start = i`

3. **Updated Global Maximum Tracking:**
   - When we find a new global maximum, update both `start` and `end` indices

**How It Works:**

```python
# Original decision (simplified):
max_ending_here = max(nums[i], max_ending_here + nums[i])

# Enhanced decision (with index tracking):
if max_ending_here + nums[i] > nums[i]:
    # Extend existing subarray
    max_ending_here = max_ending_here + nums[i]
    # temp_start remains unchanged
else:
    # Start new subarray from current position
    max_ending_here = nums[i]
    temp_start = i  # Update start of current subarray
```

**Why This Works:**
- `temp_start` tracks where the current subarray started
- When we find a better global maximum, we update `start` and `end` to point to this subarray
- The algorithm still maintains O(n) time complexity!


````python
# Let's trace through the enhanced algorithm step by step
# Example: nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

def max_subarray_kadane_traced(nums):
    """
    Enhanced Kadane's Algorithm with detailed tracing
    """
    n = len(nums)
    
    # Initialize variables
    max_ending_here = nums[0]
    max_so_far = nums[0]
    start = 0
    end = 0
    temp_start = 0
    
    print(f"Input array: {nums}")
    print(f"Initial: max_ending_here = {max_ending_here}, max_so_far = {max_so_far}")
    print(f"         start = {start}, end = {end}, temp_start = {temp_start}")
    print("\n" + "="*80)
    
    for i in range(1, n):
        print(f"\nStep {i}: Processing nums[{i}] = {nums[i]}")
        
        # Decision logic
        if max_ending_here + nums[i] > nums[i]:
            print(f"  Decision: EXTEND subarray")
            print(f"  {max_ending_here} + {nums[i]} = {max_ending_here + nums[i]} > {nums[i]}")
            max_ending_here = max_ending_here + nums[i]
            print(f"  max_ending_here = {max_ending_here} (temp_start stays {temp_start})")
        else:
            print(f"  Decision: START FRESH")
            print(f"  {max_ending_here} + {nums[i]} = {max_ending_here + nums[i]} <= {nums[i]}")
            max_ending_here = nums[i]
            temp_start = i
            print(f"  max_ending_here = {max_ending_here}, temp_start = {temp_start}")
        
        # Update global maximum
        if max_ending_here > max_so_far:
            print(f"  ✓ NEW GLOBAL MAX!")
            print(f"  {max_ending_here} > {max_so_far}")
            max_so_far = max_ending_here
            start = temp_start
            end = i
            print(f"  Updated: max_so_far = {max_so_far}, start = {start}, end = {end}")
        else:
            print(f"  No change: {max_ending_here} <= {max_so_far}")
    
    print(f"\n" + "="*80)
    print(f"Final Result:")
    print(f"  Max sum = {max_so_far}")
    print(f"  Subarray indices = [{start}:{end}]")
    print(f"  Subarray elements = {nums[start:end+1]}")
    
    return max_so_far, start, end

# Test with the main example
print("TRACING ENHANCED KADANE'S ALGORITHM")
print("="*80)
test_array = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
result_sum, result_start, result_end = max_subarray_kadane_traced(test_array)
````

## Problem 3: Longest Substring Without Repeating Characters

**Problem Statement**: Find the length of the longest substring without repeating characters.

**Example**:
- Input: s = "abcabcbb"
- Output: 3
- Explanation: The longest substring without repeating characters is "abc", with length 3.

**More Examples**:
- Input: s = "bbbbb" → Output: 1 (substring "b")
- Input: s = "pwwkew" → Output: 3 (substring "wke")
- Input: s = "" → Output: 0 (empty string)

**Constraints**:
- 0 ≤ s.length ≤ 5 × 10⁴
- s consists of English letters, digits, symbols and spaces

**Follow-up**: Can you solve it in O(n) time complexity?


### Solution 1: Brute Force Approach (Check All Substrings)

**Time Complexity**: O(n³)  
**Space Complexity**: O(min(m, n)) where m is the size of charset

**Approach**: 
- Generate all possible substrings
- For each substring, check if it has unique characters
- Keep track of the maximum length found
- This is straightforward but inefficient


````python
def longest_substring_brute_force(s):
    """
    Brute Force Solution - O(n³) time, O(min(m,n)) space
    Check all possible substrings for uniqueness
    """
    n = len(s)
    max_length = 0
    
    # Generate all possible substrings
    for i in range(n):           # Start index
        for j in range(i, n):    # End index
            # Check if substring from i to j has unique characters
            substring = s[i:j+1]
            
            # Check for uniqueness using set
            if len(substring) == len(set(substring)):
                max_length = max(max_length, len(substring))
    
    return max_length

def longest_substring_brute_force_detailed(s):
    """
    Brute Force Solution with detailed tracing
    """
    n = len(s)
    max_length = 0
    best_substring = ""
    
    print(f"Input string: '{s}'")
    print(f"String length: {n}")
    print(f"Initial max_length: {max_length}")
    print("\n" + "="*60)
    
    # Generate all possible substrings
    for i in range(n):
        print(f"\nOuter loop: i = {i}")
        for j in range(i, n):
            substring = s[i:j+1]
            unique_chars = set(substring)
            
            print(f"  Inner loop: j = {j}")
            print(f"  Checking substring: '{substring}'")
            print(f"  Length: {len(substring)}, Unique chars: {len(unique_chars)}")
            
            # Check for uniqueness
            if len(substring) == len(unique_chars):
                print(f"  ✓ UNIQUE! Length = {len(substring)}")
                if len(substring) > max_length:
                    max_length = len(substring)
                    best_substring = substring
                    print(f"  ✓ NEW MAX! Updated max_length: {max_length}")
                    print(f"  Best substring so far: '{best_substring}'")
                else:
                    print(f"  No change. max_length remains: {max_length}")
            else:
                print(f"  ✗ Has duplicates. Skipping.")
    
    print(f"\n" + "="*60)
    print(f"Final result: {max_length}")
    print(f"Best substring: '{best_substring}'")
    return max_length

# Test the brute force solution
test_cases = [
    "abcabcbb",    # Expected: 3 ("abc")
    "bbbbb",       # Expected: 1 ("b")
    "pwwkew",      # Expected: 3 ("wke")
    "",            # Expected: 0
    "a",           # Expected: 1 ("a")
    "dvdf",        # Expected: 3 ("vdf")
    "abcdef"       # Expected: 6 ("abcdef")
]

print("Brute Force Solution Results:")
for i, test_string in enumerate(test_cases):
    result = longest_substring_brute_force(test_string)
    print(f"Test {i+1}: s = '{test_string}' → {result}")
print()
````

### Solution 2: Sliding Window with Hash Set (Optimal)

**Time Complexity**: O(n)  
**Space Complexity**: O(min(m, n)) where m is the size of charset

**Approach**: 
- Use sliding window technique with two pointers
- Use a hash set to track characters in current window
- Expand window when no duplicates, shrink when duplicates found
- This is the most efficient solution


````python
def longest_substring_sliding_window(s):
    """
    Sliding Window Solution - O(n) time, O(min(m,n)) space
    Optimal solution using two pointers and hash set
    """
    n = len(s)
    if n == 0:
        return 0
    
    char_set = set()
    left = 0
    max_length = 0
    
    for right in range(n):
        # If character at 'right' is already in set, shrink window from left
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1
        
        # Add current character to set
        char_set.add(s[right])
        
        # Update maximum length
        max_length = max(max_length, right - left + 1)
    
    return max_length

def longest_substring_sliding_window_detailed(s):
    """
    Sliding Window Solution with detailed tracing
    """
    n = len(s)
    if n == 0:
        return 0
    
    char_set = set()
    left = 0
    max_length = 0
    best_substring = ""
    
    print(f"Input string: '{s}'")
    print(f"String length: {n}")
    print(f"Initial: left = {left}, max_length = {max_length}")
    print(f"         char_set = {char_set}")
    print("\n" + "="*70)
    
    for right in range(n):
        print(f"\nStep {right}: Processing s[{right}] = '{s[right]}'")
        print(f"  Current window: s[{left}:{right+1}] = '{s[left:right+1]}'")
        
        # Shrink window from left if duplicate found
        while s[right] in char_set:
            print(f"  ✗ Duplicate '{s[right]}' found in char_set {char_set}")
            print(f"  Removing s[{left}] = '{s[left]}' from set")
            char_set.remove(s[left])
            left += 1
            print(f"  Updated: left = {left}, char_set = {char_set}")
        
        # Add current character to set
        char_set.add(s[right])
        print(f"  ✓ Added '{s[right]}' to char_set: {char_set}")
        
        # Update maximum length
        current_length = right - left + 1
        if current_length > max_length:
            max_length = current_length
            best_substring = s[left:right+1]
            print(f"  ✓ NEW MAX! Length = {max_length}")
            print(f"  Best substring: '{best_substring}'")
        else:
            print(f"  No change. Current length = {current_length}, max = {max_length}")
    
    print(f"\n" + "="*70)
    print(f"Final result: {max_length}")
    print(f"Best substring: '{best_substring}'")
    return max_length

# Test the sliding window solution
print("Sliding Window Solution Results:")
for i, test_string in enumerate(test_cases):
    result = longest_substring_sliding_window(test_string)
    print(f"Test {i+1}: s = '{test_string}' → {result}")
print()
````

## Problem 4: Valid Parentheses (Stack Algorithm)

**Problem Statement**: Given a string containing only '(', ')', '{', '}', '[' and ']', determine if the input string is valid.

**Valid Conditions**:
1. Open brackets must be closed by the same type of brackets
2. Open brackets must be closed in the correct order
3. Every close bracket has a corresponding open bracket of the same type

**Examples**:
- Input: s = "()" → Output: true
- Input: s = "()[]{}" → Output: true  
- Input: s = "(]" → Output: false
- Input: s = "([)]" → Output: false
- Input: s = "{[]}" → Output: true

**Constraints**:
- 1 ≤ s.length ≤ 10⁴
- s consists of parentheses only '()[]{}'

**Follow-up**: Can you solve it in O(n) time and O(n) space?


### Solution 1: Brute Force Approach (Repeated Removal)

**Time Complexity**: O(n²)  
**Space Complexity**: O(1)

**Approach**: 
- Repeatedly remove valid pairs of parentheses
- If we can remove all characters, the string is valid
- If no more pairs can be removed but string is not empty, it's invalid
- This is simple but inefficient


````python
def valid_parentheses_brute_force(s):
    """
    Brute Force Solution - O(n²) time, O(1) space
    Repeatedly remove valid pairs until no more can be removed
    """
    while True:
        # Check if we can remove any valid pairs
        original_length = len(s)
        
        # Remove all valid pairs
        s = s.replace('()', '').replace('[]', '').replace('{}', '')
        
        # If no pairs were removed, we're done
        if len(s) == original_length:
            break
    
    # String is valid if it's empty
    return len(s) == 0

def valid_parentheses_brute_force_detailed(s):
    """
    Brute Force Solution with detailed tracing
    """
    original_s = s
    step = 0
    
    print(f"Input string: '{s}'")
    print(f"String length: {len(s)}")
    print("\n" + "="*60)
    
    while True:
        step += 1
        original_length = len(s)
        
        print(f"\nStep {step}: Current string = '{s}'")
        print(f"  Length before removal: {original_length}")
        
        # Remove all valid pairs
        s = s.replace('()', '').replace('[]', '').replace('{}', '')
        
        new_length = len(s)
        removed_pairs = (original_length - new_length) // 2
        
        print(f"  Length after removal: {new_length}")
        print(f"  Pairs removed: {removed_pairs}")
        
        # If no pairs were removed, we're done
        if new_length == original_length:
            print(f"  No more pairs can be removed. Stopping.")
            break
    
    print(f"\n" + "="*60)
    print(f"Final string: '{s}'")
    print(f"Final length: {len(s)}")
    print(f"Result: {'✓ VALID' if len(s) == 0 else '✗ INVALID'}")
    
    return len(s) == 0

# Test the brute force solution
test_cases = [
    "()",           # Expected: true
    "()[]{}",       # Expected: true
    "(]",           # Expected: false
    "([)]",         # Expected: false
    "{[]}",         # Expected: true
    "((()))",       # Expected: true
    "([{}])",       # Expected: true
    "(((",          # Expected: false
    ")))",          # Expected: false
    "",             # Expected: true (empty is valid)
    "()()",         # Expected: true
    "([)]",         # Expected: false
]

print("Brute Force Solution Results:")
for i, test_string in enumerate(test_cases):
    result = valid_parentheses_brute_force(test_string)
    print(f"Test {i+1}: s = '{test_string}' → {result}")
print()
````

### Solution 2: Stack-Based Approach (Optimal)

**Time Complexity**: O(n)  
**Space Complexity**: O(n)

**Approach**: 
- Use a stack to keep track of opening brackets
- When we see an opening bracket, push it onto the stack
- When we see a closing bracket, check if it matches the top of the stack
- If it matches, pop from stack; if not, return false
- String is valid if stack is empty at the end


````python
def valid_parentheses_stack(s):
    """
    Stack-Based Solution - O(n) time, O(n) space
    Optimal solution using stack data structure
    """
    # Create mapping of closing to opening brackets
    # ()[]{}
    # (]
    bracket_map = {')': '(', ']': '[', '}': '{'}
    
    # Stack to keep track of opening brackets
    stack = []
    
    for char in s:
        if char in bracket_map:  # Closing bracket
            # Check if stack is empty or top doesn't match
            if not stack or stack.pop() != bracket_map[char]:
                return False
        else:  # Opening bracket
            stack.append(char)
    
    # String is valid if stack is empty
    return len(stack) == 0

def valid_parentheses_stack_detailed(s):
    """
    Stack-Based Solution with detailed tracing
    """
    bracket_map = {')': '(', ']': '[', '}': '{'}
    stack = []
    
    print(f"Input string: '{s}'")
    print(f"String length: {len(s)}")
    print(f"Bracket mapping: {bracket_map}")
    print(f"Initial stack: {stack}")
    print("\n" + "="*60)
    
    for i, char in enumerate(s):
        print(f"\nStep {i+1}: Processing s[{i}] = '{char}'")
        
        if char in bracket_map:  # Closing bracket
            print(f"  '{char}' is a closing bracket")
            print(f"  Expected opening bracket: '{bracket_map[char]}'")
            
            if not stack:
                print(f"  ✗ Stack is empty! No opening bracket to match.")
                print(f"  Result: INVALID")
                return False
            elif stack[-1] != bracket_map[char]:
                print(f"  ✗ Top of stack '{stack[-1]}' doesn't match expected '{bracket_map[char]}'")
                print(f"  Result: INVALID")
                return False
            else:
                popped = stack.pop()
                print(f"  ✓ Match found! Popped '{popped}' from stack")
                print(f"  Updated stack: {stack}")
        else:  # Opening bracket
            print(f"  '{char}' is an opening bracket")
            stack.append(char)
            print(f"  Pushed '{char}' onto stack")
            print(f"  Updated stack: {stack}")
    
    print(f"\n" + "="*60)
    print(f"Final stack: {stack}")
    print(f"Stack empty: {len(stack) == 0}")
    print(f"Result: {'✓ VALID' if len(stack) == 0 else '✗ INVALID'}")
    
    return len(stack) == 0

# Test the stack-based solution
print("Stack-Based Solution Results:")
for i, test_string in enumerate(test_cases):
    result = valid_parentheses_stack(test_string)
    print(f"Test {i+1}: s = '{test_string}' → {result}")
print()
````

## Problem 5: Two Pointer Algorithm - Introduction

**What is Two Pointer Technique?**

The Two Pointer technique is an optimization approach used to solve array/string problems efficiently. Instead of using nested loops (which give O(n²) complexity), we use two pointers to traverse the data structure in a single pass (O(n) complexity).

**Key Characteristics:**
- Uses two pointers (indices) that move through the array/string
- Typically used for sorted arrays or strings
- Reduces time complexity from O(n²) to O(n)
- Often reduces space complexity as well

**Common Patterns:**
1. **Opposite Ends**: Start from both ends and move towards center
2. **Same Direction**: Both pointers start from beginning, move at different speeds
3. **Fast & Slow**: One pointer moves faster than the other (Floyd's cycle detection)

**When to Use Two Pointers:**
- ✅ Searching for pairs in sorted array
- ✅ Removing duplicates
- ✅ Palindrome checking
- ✅ Finding subarrays/substrings
- ✅ Merging sorted arrays
- ✅ Array partitioning

**Time Complexity**: O(n) - Single pass through array
**Space Complexity**: O(1) - Only using pointers, no extra space

## Problem 5.1: Two Sum (Sorted Array)

**Problem Statement**: Given a sorted array of integers and a target sum, find two numbers that add up to the target. Return their indices (1-indexed).

**Example**:
- Input: numbers = [2, 7, 11, 15], target = 9
- Output: [1, 2]
- Explanation: numbers[0] + numbers[1] = 2 + 7 = 9

**Constraints**:
- Array is sorted in non-decreasing order
- Exactly one solution exists
- You may not use the same element twice
- Return indices are 1-indexed (not 0-indexed)

**Follow-up**: Can you solve it using O(1) extra space?


### Solution 1: Brute Force Approach

**Time Complexity**: O(n²)  
**Space Complexity**: O(1)

**Approach**: 
- Use nested loops to check every pair of numbers
- Since array is sorted, we can optimize slightly but still O(n²)
- Check all pairs until we find the target sum


````python
def two_sum_sorted_brute_force(numbers, target):
    """
    Brute Force Solution - O(n²) time, O(1) space
    Check all pairs of numbers
    """
    n = len(numbers)
    
    for i in range(n):
        for j in range(i + 1, n):
            if numbers[i] + numbers[j] == target:
                return [i + 1, j + 1]  # 1-indexed
    
    return []

# Test the brute force solution
test_cases = [
    ([2, 7, 11, 15], 9),      # Expected: [1, 2]
    ([2, 3, 4], 6),            # Expected: [1, 3]
    ([-1, 0], -1),             # Expected: [1, 2]
    ([1, 2, 3, 4, 5], 9),      # Expected: [4, 5]
]

print("Brute Force Solution Results:")
for numbers, target in test_cases:
    result = two_sum_sorted_brute_force(numbers, target)
    print(f"numbers = {numbers}, target = {target} -> {result}")
    if result:
        print(f"  Verification: {numbers[result[0]-1]} + {numbers[result[1]-1]} = {target}")
    print()
````

### Solution 2: Two Pointer Approach (Optimal)

**Time Complexity**: O(n)  
**Space Complexity**: O(1)

**Approach**: 
- Use two pointers: one at the start (left) and one at the end (right)
- Since array is sorted, we can use this to our advantage
- If sum is too small, move left pointer right (increase sum)
- If sum is too large, move right pointer left (decrease sum)
- This is the most efficient solution for sorted arrays


````python
def two_sum_sorted_two_pointers(numbers, target):
    """
    Two Pointer Solution - O(n) time, O(1) space
    Optimal solution using two pointers from opposite ends
    """
    left = 0
    right = len(numbers) - 1
    
    while left < right:
        current_sum = numbers[left] + numbers[right]
        
        if current_sum == target:
            return [left + 1, right + 1]  # 1-indexed
        elif current_sum < target:
            left += 1  # Need larger sum, move left pointer right
        else:
            right -= 1  # Need smaller sum, move right pointer left
    
    return []

# Test the two pointer solution
print("Two Pointer Solution Results:")
for numbers, target in test_cases:
    result = two_sum_sorted_two_pointers(numbers, target)
    print(f"numbers = {numbers}, target = {target} -> {result}")
    if result:
        print(f"  Verification: {numbers[result[0]-1]} + {numbers[result[1]-1]} = {target}")
    print()

# Verify both solutions give same results
print("Verification - Both solutions should give identical results:")
print("=" * 60)
for numbers, target in test_cases:
    brute_result = two_sum_sorted_brute_force(numbers, target)
    two_ptr_result = two_sum_sorted_two_pointers(numbers, target)
    print(f"numbers = {numbers}, target = {target}")
    print(f"  Brute Force: {brute_result}")
    print(f"  Two Pointer: {two_ptr_result}")
    print(f"  Match: {'✓' if brute_result == two_ptr_result else '✗'}")
    print()
````

### Comparison and Analysis

| Approach | Time Complexity | Space Complexity | Pros | Cons |
|----------|----------------|------------------|------|------|
| **Brute Force** | O(n²) | O(1) | Simple to understand | Slow for large inputs |
| **Two Pointer** | O(n) | O(1) | Optimal time, single pass | Requires sorted array |

**Key Insight**: Since the array is sorted, we can eliminate half the search space at each step by moving pointers strategically.

## Problem 5.2: Remove Duplicates from Sorted Array

**Problem Statement**: Given a sorted array, remove duplicates in-place such that each element appears only once. Return the number of unique elements.

**Example**:
- Input: nums = [1, 1, 2]
- Output: 2, nums = [1, 2, _]
- Explanation: Your function should return k = 2, with the first two elements of nums being 1 and 2 respectively.

**Constraints**:
- 1 ≤ nums.length ≤ 3 × 10⁴
- -100 ≤ nums[i] ≤ 100
- Array is sorted in non-decreasing order

**Follow-up**: Can you solve it in O(1) extra space?


### Solution 1: Using Extra Space (Brute Force)

**Time Complexity**: O(n)  
**Space Complexity**: O(n)

**Approach**: 
- Create a new array/list to store unique elements
- Iterate through array and add elements that haven't been seen
- This is straightforward but uses extra space


````python
def remove_duplicates_extra_space(nums):
    """
    Solution using extra space - O(n) time, O(n) space
    Creates a new list with unique elements
    """
    if not nums:
        return 0
    
    unique = []
    seen = set()
    
    for num in nums:
        if num not in seen:
            unique.append(num)
            seen.add(num)
    
    # Copy back to original array (if needed)
    for i in range(len(unique)):
        nums[i] = unique[i]
    
    return len(unique)

# Test
test_cases_remove = [
    [1, 1, 2],
    [0, 0, 1, 1, 1, 2, 2, 3, 3, 4],
    [1, 2, 3],
    [1, 1, 1],
]

print("Remove Duplicates - Extra Space Solution:")
for test_nums in test_cases_remove:
    nums_copy = test_nums.copy()
    k = remove_duplicates_extra_space(nums_copy)
    print(f"Input: {test_nums}")
    print(f"Output: k = {k}, nums = {nums_copy[:k]}")
    print()
````

### Solution 2: Two Pointer Approach (Optimal)

**Time Complexity**: O(n)  
**Space Complexity**: O(1)

**Approach**: 
- Use two pointers: one for writing (slow) and one for reading (fast)
- Since array is sorted, duplicates are adjacent
- Move fast pointer to skip duplicates
- Write unique elements at slow pointer position
- This is the optimal in-place solution


````python
def remove_duplicates_two_pointers(nums):
    """
    Two Pointer Solution - O(n) time, O(1) space
    Optimal in-place solution using two pointers
    """
    if not nums:
        return 0
    
    # Slow pointer: position to write next unique element
    # Fast pointer: position to read next element
    slow = 0
    
    for fast in range(1, len(nums)):
        # If current element is different from previous unique element
        if nums[fast] != nums[slow]:
            slow += 1
            nums[slow] = nums[fast]
    
    return slow + 1  # Return count of unique elements

# Test the two pointer solution
print("Remove Duplicates - Two Pointer Solution:")
for test_nums in test_cases_remove:
    nums_copy = test_nums.copy()
    k = remove_duplicates_two_pointers(nums_copy)
    print(f"Input: {test_nums}")
    print(f"Output: k = {k}, nums = {nums_copy[:k]}")
    print()

# Verify both solutions
print("Verification:")
print("=" * 60)
for test_nums in test_cases_remove:
    copy1 = test_nums.copy()
    copy2 = test_nums.copy()
    k1 = remove_duplicates_extra_space(copy1)
    k2 = remove_duplicates_two_pointers(copy2)
    print(f"Input: {test_nums}")
    print(f"  Extra Space: k = {k1}, nums = {copy1[:k1]}")
    print(f"  Two Pointer: k = {k2}, nums = {copy2[:k2]}")
    print(f"  Match: {'✓' if k1 == k2 and copy1[:k1] == copy2[:k2] else '✗'}")
    print()
````

## Problem 5.3: Valid Palindrome

**Problem Statement**: Given a string, determine if it is a palindrome, considering only alphanumeric characters and ignoring cases.

**Example**:
- Input: s = "A man, a plan, a canal: Panama"
- Output: true
- Explanation: "amanaplanacanalpanama" is a palindrome.

**More Examples**:
- Input: s = "race a car" → Output: false
- Input: s = " " → Output: true (empty string is palindrome)

**Constraints**:
- 1 ≤ s.length ≤ 2 × 10⁵
- s consists only of printable ASCII characters

**Follow-up**: Can you solve it using O(1) extra space?


### Solution 1: String Reversal (Brute Force)

**Time Complexity**: O(n)  
**Space Complexity**: O(n)

**Approach**: 
- Clean the string (remove non-alphanumeric, convert to lowercase)
- Reverse the cleaned string
- Compare with original cleaned string
- Simple but uses extra space


````python
def is_palindrome_brute_force(s):
    """
    Brute Force Solution - O(n) time, O(n) space
    Clean string and compare with reversed version
    """
    # Clean the string: keep only alphanumeric characters and convert to lowercase
    cleaned = ''.join(char.lower() for char in s if char.isalnum())
    
    # Compare with reversed version
    return cleaned == cleaned[::-1]

# Test
test_cases_palindrome = [
    "A man, a plan, a canal: Panama",
    "race a car",
    " ",
    "Madam",
    "No 'x' in Nixon",
    "a",
]

print("Valid Palindrome - Brute Force Solution:")
for test_str in test_cases_palindrome:
    result = is_palindrome_brute_force(test_str)
    print(f"Input: '{test_str}' → {result}")
print()
````

### Solution 2: Two Pointer Approach (Optimal)

**Time Complexity**: O(n)  
**Space Complexity**: O(1)

**Approach**: 
- Use two pointers: one from start (left) and one from end (right)
- Skip non-alphanumeric characters
- Compare characters at both pointers (case-insensitive)
- Move pointers towards center until they meet
- This is the optimal O(1) space solution


````python
def is_palindrome_two_pointers(s):
    """
    Two Pointer Solution - O(n) time, O(1) space
    Optimal solution using two pointers from opposite ends
    """
    left = 0
    right = len(s) - 1
    
    while left < right:
        # Skip non-alphanumeric characters from left
        while left < right and not s[left].isalnum():
            left += 1
        
        # Skip non-alphanumeric characters from right
        while left < right and not s[right].isalnum():
            right -= 1
        
        # Compare characters (case-insensitive)
        if s[left].lower() != s[right].lower():
            return False
        
        left += 1
        right -= 1
    
    return True

# Test the two pointer solution
print("Valid Palindrome - Two Pointer Solution:")
for test_str in test_cases_palindrome:
    result = is_palindrome_two_pointers(test_str)
    print(f"Input: '{test_str}' → {result}")
print()

# Verify both solutions
print("Verification:")
print("=" * 60)
for test_str in test_cases_palindrome:
    brute_result = is_palindrome_brute_force(test_str)
    two_ptr_result = is_palindrome_two_pointers(test_str)
    print(f"Input: '{test_str}'")
    print(f"  Brute Force: {brute_result}")
    print(f"  Two Pointer: {two_ptr_result}")
    print(f"  Match: {'✓' if brute_result == two_ptr_result else '✗'}")
    print()
````

## Problem 5.4: Container With Most Water

**Problem Statement**: Given an array of heights, find two lines that together with the x-axis form a container, such that the container contains the most water. Return the maximum area.

**Example**:
- Input: height = [1,8,6,2,5,4,8,3,7]
- Output: 49
- Explanation: The max area is between indices 1 and 8: min(8,7) × (8-1) = 7 × 7 = 49

**Visual Example**:
```
     |
     |     |
     |  |  |
  |  |  |  |  |
|_|__|__|__|__|__
1 8 6 2 5 4 8 3 7
```

**Constraints**:
- n == height.length
- 2 ≤ n ≤ 10⁵
- 0 ≤ height[i] ≤ 10⁴

**Follow-up**: Can you solve it in O(n) time?


### Solution 1: Brute Force (Check All Pairs)

**Time Complexity**: O(n²)  
**Space Complexity**: O(1)

**Approach**: 
- Check all possible pairs of lines
- Calculate area for each pair: area = min(height[i], height[j]) × (j - i)
- Keep track of maximum area
- Simple but inefficient for large inputs


````python
def max_area_brute_force(height):
    """
    Brute Force Solution - O(n²) time, O(1) space
    Check all possible pairs of lines
    """
    max_area = 0
    n = len(height)
    
    for i in range(n):
        for j in range(i + 1, n):
            # Area = min(height[i], height[j]) × width
            area = min(height[i], height[j]) * (j - i)
            max_area = max(max_area, area)
    
    return max_area

# Test
test_cases_container = [
    [1, 8, 6, 2, 5, 4, 8, 3, 7],  # Expected: 49
    [1, 1],                         # Expected: 1
    [4, 3, 2, 1, 4],                # Expected: 16
    [1, 2, 1],                      # Expected: 2
]

print("Container With Most Water - Brute Force Solution:")
for test_heights in test_cases_container:
    result = max_area_brute_force(test_heights)
    print(f"height = {test_heights} → max_area = {result}")
print()
````

### Solution 2: Two Pointer Approach (Optimal)

**Time Complexity**: O(n)  
**Space Complexity**: O(1)

**Approach**: 
- Use two pointers: one at start (left) and one at end (right)
- Calculate area: min(height[left], height[right]) × (right - left)
- Move the pointer with smaller height towards center
- Why? Because moving the taller pointer can only decrease area (width decreases, height can't increase)
- This is the optimal O(n) solution


````python
def max_area_two_pointers(height):
    """
    Two Pointer Solution - O(n) time, O(1) space
    Optimal solution using two pointers from opposite ends
    """
    left = 0
    right = len(height) - 1
    max_area = 0
    
    while left < right:
        # Calculate current area
        width = right - left
        current_area = min(height[left], height[right]) * width
        max_area = max(max_area, current_area)
        
        # Move pointer with smaller height
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
    
    return max_area

# Test the two pointer solution
print("Container With Most Water - Two Pointer Solution:")
for test_heights in test_cases_container:
    result = max_area_two_pointers(test_heights)
    print(f"height = {test_heights} → max_area = {result}")
print()

# Verify both solutions
print("Verification:")
print("=" * 60)
for test_heights in test_cases_container:
    brute_result = max_area_brute_force(test_heights)
    two_ptr_result = max_area_two_pointers(test_heights)
    print(f"height = {test_heights}")
    print(f"  Brute Force: {brute_result}")
    print(f"  Two Pointer: {two_ptr_result}")
    print(f"  Match: {'✓' if brute_result == two_ptr_result else '✗'}")
    print()
````

## Problem 5.5: Move Zeroes

**Problem Statement**: Given an array, move all 0's to the end while maintaining the relative order of non-zero elements. You must do this in-place without making a copy of the array.

**Example**:
- Input: nums = [0,1,0,3,12]
- Output: [1,3,12,0,0]
- Explanation: All zeros moved to end, non-zeros maintain relative order

**Constraints**:
- 1 ≤ nums.length ≤ 10⁴
- -2³¹ ≤ nums[i] ≤ 2³¹ - 1

**Follow-up**: Can you minimize the total number of operations?


### Solution 1: Using Extra Array (Brute Force)

**Time Complexity**: O(n)  
**Space Complexity**: O(n)

**Approach**: 
- Create a new array
- First add all non-zero elements
- Then fill remaining positions with zeros
- Copy back to original array
- Simple but uses extra space


````python
def move_zeroes_extra_space(nums):
    """
    Solution using extra space - O(n) time, O(n) space
    Creates new array and copies back
    """
    result = []
    
    # Add all non-zero elements
    for num in nums:
        if num != 0:
            result.append(num)
    
    # Fill remaining positions with zeros
    while len(result) < len(nums):
        result.append(0)
    
    # Copy back to original array
    for i in range(len(nums)):
        nums[i] = result[i]

# Test
test_cases_zeroes = [
    [0, 1, 0, 3, 12],
    [0],
    [1, 0, 2, 0, 3],
    [0, 0, 0, 1],
    [1, 2, 3],
]

print("Move Zeroes - Extra Space Solution:")
for test_nums in test_cases_zeroes:
    nums_copy = test_nums.copy()
    move_zeroes_extra_space(nums_copy)
    print(f"Input: {test_nums}")
    print(f"Output: {nums_copy}")
    print()
````

### Solution 2: Two Pointer Approach (Optimal)

**Time Complexity**: O(n)  
**Space Complexity**: O(1)

**Approach**: 
- Use two pointers: one for writing (slow) and one for reading (fast)
- Slow pointer tracks position for next non-zero element
- Fast pointer scans through array
- When we find non-zero, swap with slow pointer position
- This minimizes operations and uses O(1) space


````python
def move_zeroes_two_pointers(nums):
    """
    Two Pointer Solution - O(n) time, O(1) space
    Optimal in-place solution using two pointers
    """
    # Slow pointer: position to write next non-zero element
    slow = 0
    
    # Fast pointer: scans through array
    for fast in range(len(nums)):
        if nums[fast] != 0:
            # Swap non-zero element to slow pointer position
            nums[slow], nums[fast] = nums[fast], nums[slow]
            slow += 1
    
    # All zeros are automatically at the end after slow pointer

# Test the two pointer solution
print("Move Zeroes - Two Pointer Solution:")
for test_nums in test_cases_zeroes:
    nums_copy = test_nums.copy()
    move_zeroes_two_pointers(nums_copy)
    print(f"Input: {test_nums}")
    print(f"Output: {nums_copy}")
    print()

# Verify both solutions
print("Verification:")
print("=" * 60)
for test_nums in test_cases_zeroes:
    copy1 = test_nums.copy()
    copy2 = test_nums.copy()
    move_zeroes_extra_space(copy1)
    move_zeroes_two_pointers(copy2)
    print(f"Input: {test_nums}")
    print(f"  Extra Space: {copy1}")
    print(f"  Two Pointer: {copy2}")
    print(f"  Match: {'✓' if copy1 == copy2 else '✗'}")
    print()
````

## Two Pointer Algorithm - Summary & Interview Tips

### Quick Reference Guide

| Problem | Pattern | Time Complexity | Space Complexity |
|---------|---------|----------------|------------------|
| Two Sum (Sorted) | Opposite Ends | O(n) | O(1) |
| Remove Duplicates | Same Direction (Slow/Fast) | O(n) | O(1) |
| Valid Palindrome | Opposite Ends | O(n) | O(1) |
| Container With Most Water | Opposite Ends | O(n) | O(1) |
| Move Zeroes | Same Direction (Slow/Fast) | O(n) | O(1) |

### Common Two Pointer Patterns

**1. Opposite Ends Pattern:**
```python
left = 0
right = len(arr) - 1

while left < right:
    # Process elements at left and right
    # Move pointers based on condition
    if condition:
        left += 1
    else:
        right -= 1
```
**Use Cases**: Searching pairs, palindrome checking, container problems

**2. Same Direction (Slow/Fast) Pattern:**
```python
slow = 0
for fast in range(len(arr)):
    if should_keep(arr[fast]):
        arr[slow] = arr[fast]
        slow += 1
```
**Use Cases**: Removing elements, partitioning arrays, in-place modifications

**3. Fast & Slow (Floyd's Cycle) Pattern:**
```python
slow = fast = head
while fast and fast.next:
    slow = slow.next
    fast = fast.next.next
    if slow == fast:
        # Cycle detected
```
**Use Cases**: Cycle detection, finding middle element

### Interview Strategy for Two Pointer Problems

**Step 1: Identify the Pattern**
- ✅ Are we looking for pairs? → Opposite Ends
- ✅ Are we removing/filtering elements? → Same Direction
- ✅ Are we checking for cycles? → Fast & Slow

**Step 2: Start with Brute Force**
- Always mention the O(n²) brute force solution first
- Shows you understand the problem
- Demonstrates problem-solving approach

**Step 3: Optimize with Two Pointers**
- Explain why two pointers work
- Walk through an example
- Discuss time/space complexity improvements

**Step 4: Handle Edge Cases**
- Empty arrays/strings
- Single element
- All elements same
- No solution found

### Key Insights to Remember

1. **Sorted Arrays**: Two pointers work great because we can eliminate half the search space
2. **Opposite Ends**: Use when we need to find pairs or compare from both ends
3. **Same Direction**: Use when we need to filter/transform in-place
4. **Greedy Approach**: Two pointers often use greedy decisions (move pointer that gives better result)

- ❌ Not updating both pointers correctly
