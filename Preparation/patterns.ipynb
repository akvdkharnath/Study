## Strategy Pattern

### What is Strategy Pattern?

The Strategy Pattern is a behavioral design pattern that enables selecting an algorithm's behavior at runtime. Instead of implementing a single algorithm directly, code receives run-time instructions on which algorithm to use.

**Key Concept**: Define a family of algorithms, encapsulate each one, and make them interchangeable.


### Real-World Problem: Payment Processing System

Imagine you're building an e-commerce application that needs to process payments. Different customers want to pay using different methods:
- Credit Card
- PayPal
- Bank Transfer
- Cryptocurrency

**Without Strategy Pattern**: You'd have a lot of if-else statements or switch cases, making the code hard to maintain and extend.


## Code BEFORE Strategy Pattern (Problem)

Let's see how the code looks without using the Strategy pattern:


````python
# ❌ BAD CODE: Without Strategy Pattern

class PaymentProcessor:
    """Payment processor without Strategy pattern - lots of if-else statements"""
    
    def process_payment(self, amount, payment_method):
        """Process payment based on payment method"""
        
        if payment_method == "credit_card":
            # Credit card processing logic
            print(f"Processing ${amount} via Credit Card")
            print("  - Validating card details...")
            print("  - Charging credit card...")
            print("  - Transaction completed!")
            return True
            
        elif payment_method == "paypal":
            # PayPal processing logic
            print(f"Processing ${amount} via PayPal")
            print("  - Redirecting to PayPal...")
            print("  - Verifying PayPal account...")
            print("  - Transaction completed!")
            return True
            
        elif payment_method == "bank_transfer":
            # Bank transfer processing logic
            print(f"Processing ${amount} via Bank Transfer")
            print("  - Initiating bank transfer...")
            print("  - Verifying bank account...")
            print("  - Transaction completed!")
            return True
            
        elif payment_method == "crypto":
            # Cryptocurrency processing logic
            print(f"Processing ${amount} via Cryptocurrency")
            print("  - Connecting to blockchain...")
            print("  - Verifying wallet address...")
            print("  - Transaction completed!")
            return True
            
        else:
            print(f"Unknown payment method: {payment_method}")
            return False


# Usage
processor = PaymentProcessor()

# Process different payments
processor.process_payment(100, "credit_card")
print()
processor.process_payment(200, "paypal")
print()
processor.process_payment(150, "bank_transfer")
print()
processor.process_payment(300, "crypto")
````

### Problems with the Above Code:

1. **Violates Open/Closed Principle**: To add a new payment method, you need to modify the existing class
2. **Hard to Test**: Each payment method is tightly coupled
3. **Code Duplication**: Similar structure repeated for each method
4. **Difficult to Maintain**: All logic in one place
5. **Not Flexible**: Can't easily swap payment methods at runtime


## Code AFTER Strategy Pattern (Solution)

Now let's refactor using the Strategy pattern:


````python
# ✅ GOOD CODE: With Strategy Pattern

from abc import ABC, abstractmethod

# Step 1: Define the Strategy Interface
class PaymentStrategy(ABC):
    """Abstract base class for payment strategies"""
    
    @abstractmethod
    def process_payment(self, amount):
        """Process payment - must be implemented by concrete strategies"""
        pass


# Step 2: Implement Concrete Strategies
class CreditCardPayment(PaymentStrategy):
    """Concrete strategy for Credit Card payments"""
    
    def process_payment(self, amount):
        print(f"Processing ${amount} via Credit Card")
        print("  - Validating card details...")
        print("  - Charging credit card...")
        print("  - Transaction completed!")
        return True


class PayPalPayment(PaymentStrategy):
    """Concrete strategy for PayPal payments"""
    
    def process_payment(self, amount):
        print(f"Processing ${amount} via PayPal")
        print("  - Redirecting to PayPal...")
        print("  - Verifying PayPal account...")
        print("  - Transaction completed!")
        return True


class BankTransferPayment(PaymentStrategy):
    """Concrete strategy for Bank Transfer payments"""
    
    def process_payment(self, amount):
        print(f"Processing ${amount} via Bank Transfer")
        print("  - Initiating bank transfer...")
        print("  - Verifying bank account...")
        print("  - Transaction completed!")
        return True


class CryptocurrencyPayment(PaymentStrategy):
    """Concrete strategy for Cryptocurrency payments"""
    
    def process_payment(self, amount):
        print(f"Processing ${amount} via Cryptocurrency")
        print("  - Connecting to blockchain...")
        print("  - Verifying wallet address...")
        print("  - Transaction completed!")
        return True


# Step 3: Context Class that uses the Strategy
class PaymentProcessor:
    """Context class that uses payment strategies"""
    
    def __init__(self, payment_strategy: PaymentStrategy):
        """Initialize with a payment strategy"""
        self._payment_strategy = payment_strategy
    
    def set_payment_strategy(self, payment_strategy: PaymentStrategy):
        """Change payment strategy at runtime"""
        self._payment_strategy = payment_strategy
    
    def process_payment(self, amount):
        """Process payment using the current strategy"""
        return self._payment_strategy.process_payment(amount)


# Usage
print("=== Using Strategy Pattern ===\n")

# Create processor with Credit Card strategy
processor = PaymentProcessor(CreditCardPayment())
processor.process_payment(100)
print()

# Change strategy to PayPal
processor.set_payment_strategy(PayPalPayment())
processor.process_payment(200)
print()

# Change strategy to Bank Transfer
processor.set_payment_strategy(BankTransferPayment())
processor.process_payment(150)
print()

# Change strategy to Cryptocurrency
processor.set_payment_strategy(CryptocurrencyPayment())
processor.process_payment(300)
````

## More Realistic Example: E-commerce Checkout

Let's see a more complete example with actual payment processing:


````python
# More Realistic Example with Strategy Pattern

from abc import ABC, abstractmethod
from typing import Dict

class PaymentStrategy(ABC):
    """Payment strategy interface"""
    
    @abstractmethod
    def pay(self, amount: float, payment_info: Dict) -> bool:
        """Process payment"""
        pass
    
    @abstractmethod
    def get_payment_method_name(self) -> str:
        """Return payment method name"""
        pass


class CreditCardStrategy(PaymentStrategy):
    """Credit Card payment strategy"""
    
    def pay(self, amount: float, payment_info: Dict) -> bool:
        card_number = payment_info.get('card_number')
        cvv = payment_info.get('cvv')
        
        print(f"💳 Processing ${amount} via Credit Card")
        print(f"   Card: ****{card_number[-4:]}")
        
        # Simulate validation
        if len(card_number) == 16 and len(cvv) == 3:
            print("   ✅ Payment successful!")
            return True
        else:
            print("   ❌ Payment failed: Invalid card details")
            return False
    
    def get_payment_method_name(self) -> str:
        return "Credit Card"


class PayPalStrategy(PaymentStrategy):
    """PayPal payment strategy"""
    
    def pay(self, amount: float, payment_info: Dict) -> bool:
        email = payment_info.get('email')
        
        print(f"🅿️ Processing ${amount} via PayPal")
        print(f"   Email: {email}")
        
        # Simulate validation
        if '@' in email:
            print("   ✅ Payment successful!")
            return True
        else:
            print("   ❌ Payment failed: Invalid email")
            return False
    
    def get_payment_method_name(self) -> str:
        return "PayPal"


class BankTransferStrategy(PaymentStrategy):
    """Bank Transfer payment strategy"""
    
    def pay(self, amount: float, payment_info: Dict) -> bool:
        account_number = payment_info.get('account_number')
        routing_number = payment_info.get('routing_number')
        
        print(f"🏦 Processing ${amount} via Bank Transfer")
        print(f"   Account: ****{account_number[-4:]}")
        
        # Simulate validation
        if len(account_number) >= 8 and len(routing_number) == 9:
            print("   ✅ Payment successful!")
            return True
        else:
            print("   ❌ Payment failed: Invalid account details")
            return False
    
    def get_payment_method_name(self) -> str:
        return "Bank Transfer"


class ShoppingCart:
    """Shopping cart that uses payment strategy"""
    
    def __init__(self):
        self.items = []
        self.total = 0.0
    
    def add_item(self, item: str, price: float):
        """Add item to cart"""
        self.items.append(item)
        self.total += price
        print(f"Added {item} - ${price}")
    
    def checkout(self, payment_strategy: PaymentStrategy, payment_info: Dict):
        """Checkout using selected payment strategy"""
        print(f"\n{'='*50}")
        print(f"Checkout Summary")
        print(f"{'='*50}")
        print(f"Items: {', '.join(self.items)}")
        print(f"Total: ${self.total}")
        print(f"Payment Method: {payment_strategy.get_payment_method_name()}")
        print(f"{'='*50}\n")
        
        return payment_strategy.pay(self.total, payment_info)


# Usage Example
cart = ShoppingCart()
cart.add_item("Laptop", 999.99)
cart.add_item("Mouse", 29.99)
cart.add_item("Keyboard", 79.99)

print("\n" + "="*50)
print("Example 1: Paying with Credit Card")
print("="*50)
credit_card_strategy = CreditCardStrategy()
payment_info = {
    'card_number': '1234567890123456',
    'cvv': '123'
}
cart.checkout(credit_card_strategy, payment_info)

print("\n" + "="*50)
print("Example 2: Paying with PayPal")
print("="*50)
paypal_strategy = PayPalStrategy()
payment_info = {
    'email': 'user@example.com'
}
cart.checkout(paypal_strategy, payment_info)
````

## Key Benefits of Strategy Pattern

### 1. **Open/Closed Principle**
- Open for extension (add new strategies)
- Closed for modification (don't change existing code)

### 2. **Separation of Concerns**
- Each strategy is independent
- Easy to test individual strategies

### 3. **Runtime Flexibility**
- Can change strategy at runtime
- No need to modify context class

### 4. **Eliminates Conditional Statements**
- No more if-else chains
- Cleaner, more maintainable code

### 5. **Easy to Add New Strategies**
- Just create a new class implementing the interface
- No need to touch existing code


## When to Use Strategy Pattern

Use Strategy Pattern when:

1. ✅ **Multiple ways to perform a task**: Different algorithms for the same operation
2. ✅ **Want to avoid if-else chains**: Replace conditional logic with polymorphism
3. ✅ **Need runtime algorithm selection**: Choose algorithm at runtime
4. ✅ **Want to isolate algorithm details**: Hide implementation from client
5. ✅ **Need to add new algorithms frequently**: Easy to extend without modifying existing code

### Common Use Cases:
- Payment processing (as shown above)
- Sorting algorithms (QuickSort, MergeSort, BubbleSort)
- Compression algorithms (ZIP, RAR, 7ZIP)
- Validation strategies (Email, Phone, Credit Card validation)
- Discount calculation (Student, Senior, Regular customer)
- File export formats (PDF, CSV, Excel)


## Summary: Strategy Pattern

### Structure:

```
┌─────────────────────┐
│  Strategy (ABC)     │  ← Interface
│  + execute()        │
└─────────────────────┘
         ▲
         │ implements
         │
    ┌────┴────┬──────────┬──────────┐
    │         │          │          │
┌───┴───┐ ┌──┴───┐  ┌───┴───┐  ┌───┴───┐
│Concrete│ │Concrete│ │Concrete│ │Concrete│
│Strategy│ │Strategy│ │Strategy│ │Strategy│
│   1    │ │   2    │ │   3    │ │   N    │
└───┬───┘ └───┬───┘ └───┬───┘ └───┬───┘
    │         │          │          │
    └─────────┴──────────┴──────────┘
              │ uses
              ▼
    ┌─────────────────┐
    │   Context       │  ← Uses strategy
    │   + strategy    │
    │   + execute()    │
    └─────────────────┘
```

### Key Components:

1. **Strategy Interface**: Abstract base class defining the algorithm interface
2. **Concrete Strategies**: Different implementations of the algorithm
3. **Context**: Class that uses the strategy (has a reference to strategy)

### Advantages:

✅ **Flexibility**: Easy to add new strategies  
✅ **Maintainability**: Each strategy is isolated  
✅ **Testability**: Easy to test each strategy independently  
✅ **No Conditionals**: Eliminates if-else chains  
✅ **Runtime Selection**: Choose algorithm at runtime  

### Disadvantages:

❌ **More Classes**: Increases number of classes  
❌ **Client Awareness**: Client must know about different strategies  
❌ **Overhead**: Slight performance overhead due to indirection  

### Remember:

> **"Strategy Pattern lets you define a family of algorithms, put each of them into a separate class, and make their objects interchangeable."**

## Decorator Pattern

### What is Decorator Pattern?

The Decorator Pattern is a structural design pattern that allows you to attach new behaviors to objects by placing these objects inside special wrapper objects that contain the behaviors.

**Key Concept**: Add new functionality to objects dynamically without altering their structure. It's like adding layers to an object.

**Real-world analogy**: Think of a pizza - you start with a base pizza, then add toppings (decorators) like cheese, pepperoni, mushrooms, etc. Each topping adds something new without changing the base pizza.


### Real-World Problem: Coffee Shop Ordering System

Imagine you're building a coffee shop application. Customers can order:
- Base coffee (Espresso, Latte, Cappuccino)
- Add-ons (Milk, Sugar, Whipped Cream, Caramel, Chocolate)

**Without Decorator Pattern**: You'd need a class for every possible combination:
- Espresso
- EspressoWithMilk
- EspressoWithSugar
- EspressoWithMilkAndSugar
- EspressoWithMilkAndSugarAndWhippedCream
- ... (exponential growth!)

**With Decorator Pattern**: You can dynamically add features to any coffee.


## Code BEFORE Decorator Pattern (Problem)

Let's see how the code looks without using the Decorator pattern:


````python
# ❌ BAD CODE: Without Decorator Pattern

class Coffee:
    """Base coffee class"""
    def __init__(self):
        self.description = "Coffee"
        self.cost = 2.0
    
    def get_description(self):
        return self.description
    
    def get_cost(self):
        return self.cost


# Problem: Need separate classes for every combination!
class Espresso:
    def __init__(self):
        self.description = "Espresso"
        self.cost = 2.5

class EspressoWithMilk:
    def __init__(self):
        self.description = "Espresso with Milk"
        self.cost = 2.5 + 0.5

class EspressoWithSugar:
    def __init__(self):
        self.description = "Espresso with Sugar"
        self.cost = 2.5 + 0.2

class EspressoWithMilkAndSugar:
    def __init__(self):
        self.description = "Espresso with Milk and Sugar"
        self.cost = 2.5 + 0.5 + 0.2

class EspressoWithMilkAndSugarAndWhippedCream:
    def __init__(self):
        self.description = "Espresso with Milk, Sugar, and Whipped Cream"
        self.cost = 2.5 + 0.5 + 0.2 + 0.7

# ... and so on for EVERY possible combination!

# Usage
print("=== Without Decorator Pattern ===\n")

espresso = Espresso()
print(f"{espresso.description}: ${espresso.cost}")

espresso_milk = EspressoWithMilk()
print(f"{espresso_milk.description}: ${espresso_milk.cost}")

espresso_milk_sugar = EspressoWithMilkAndSugar()
print(f"{espresso_milk_sugar.description}: ${espresso_milk_sugar.cost}")

# Problem: What if customer wants Espresso + Milk + Sugar + Caramel + Chocolate?
# We need to create ANOTHER class! This is not scalable!
````

### Problems with the Above Code:

1. **Exponential Class Explosion**: Need a class for every combination (2^n combinations for n add-ons)
2. **Code Duplication**: Similar code repeated across classes
3. **Hard to Maintain**: Adding a new add-on requires creating many new classes
4. **Not Flexible**: Can't add/remove features dynamically
5. **Violates Open/Closed Principle**: Must modify existing code to add new combinations


## Code AFTER Decorator Pattern (Solution)

Now let's refactor using the Decorator pattern:


````python
# ✅ GOOD CODE: With Decorator Pattern

from abc import ABC, abstractmethod

# Step 1: Define Component Interface
class Coffee(ABC):
    """Abstract component - base coffee interface"""
    
    @abstractmethod
    def get_description(self) -> str:
        """Get coffee description"""
        pass
    
    @abstractmethod
    def get_cost(self) -> float:
        """Get coffee cost"""
        pass


# Step 2: Create Concrete Component
class Espresso(Coffee):
    """Concrete component - base coffee"""
    
    def get_description(self) -> str:
        return "Espresso"
    
    def get_cost(self) -> float:
        return 2.5


class Latte(Coffee):
    """Concrete component - another base coffee"""
    
    def get_description(self) -> str:
        return "Latte"
    
    def get_cost(self) -> float:
        return 3.0


# Step 3: Create Decorator Base Class
class CoffeeDecorator(Coffee):
    """Base decorator class"""
    
    def __init__(self, coffee: Coffee):
        self._coffee = coffee
    
    def get_description(self) -> str:
        return self._coffee.get_description()
    
    def get_cost(self) -> float:
        return self._coffee.get_cost()


# Step 4: Create Concrete Decorators
class MilkDecorator(CoffeeDecorator):
    """Concrete decorator - adds milk"""
    
    def get_description(self) -> str:
        return self._coffee.get_description() + ", Milk"
    
    def get_cost(self) -> float:
        return self._coffee.get_cost() + 0.5


class SugarDecorator(CoffeeDecorator):
    """Concrete decorator - adds sugar"""
    
    def get_description(self) -> str:
        return self._coffee.get_description() + ", Sugar"
    
    def get_cost(self) -> float:
        return self._coffee.get_cost() + 0.2


class WhippedCreamDecorator(CoffeeDecorator):
    """Concrete decorator - adds whipped cream"""
    
    def get_description(self) -> str:
        return self._coffee.get_description() + ", Whipped Cream"
    
    def get_cost(self) -> float:
        return self._coffee.get_cost() + 0.7


class CaramelDecorator(CoffeeDecorator):
    """Concrete decorator - adds caramel"""
    
    def get_description(self) -> str:
        return self._coffee.get_description() + ", Caramel"
    
    def get_cost(self) -> float:
        return self._coffee.get_cost() + 0.6


class ChocolateDecorator(CoffeeDecorator):
    """Concrete decorator - adds chocolate"""
    
    def get_description(self) -> str:
        return self._coffee.get_description() + ", Chocolate"
    
    def get_cost(self) -> float:
        return self._coffee.get_cost() + 0.8


# Usage
print("=== Using Decorator Pattern ===\n")

# Simple espresso
coffee1 = Espresso()
print(f"{coffee1.get_description()}: ${coffee1.get_cost()}")

# Espresso with milk
coffee2 = MilkDecorator(Espresso())
print(f"{coffee2.get_description()}: ${coffee2.get_cost()}")

# Espresso with milk and sugar
coffee3 = SugarDecorator(MilkDecorator(Espresso()))
print(f"{coffee3.get_description()}: ${coffee3.get_cost()}")

# Espresso with multiple add-ons
coffee4 = ChocolateDecorator(
    CaramelDecorator(
        WhippedCreamDecorator(
            SugarDecorator(
                MilkDecorator(Espresso())
            )
        )
    )
)
print(f"{coffee4.get_description()}: ${coffee4.get_cost()}")

# Latte with caramel
coffee5 = CaramelDecorator(Latte())
print(f"{coffee5.get_description()}: ${coffee5.get_cost()}")
````

## More Realistic Example: Text Formatting

Let's see another practical example with text formatting decorators:


````python
# Example: Text Formatting with Decorators

from abc import ABC, abstractmethod

class TextComponent(ABC):
    """Component interface for text"""
    
    @abstractmethod
    def render(self) -> str:
        """Render the text"""
        pass


class PlainText(TextComponent):
    """Concrete component - plain text"""
    
    def __init__(self, text: str):
        self.text = text
    
    def render(self) -> str:
        return self.text


class TextDecorator(TextComponent):
    """Base decorator for text"""
    
    def __init__(self, component: TextComponent):
        self._component = component
    
    def render(self) -> str:
        return self._component.render()


class BoldDecorator(TextDecorator):
    """Decorator that makes text bold"""
    
    def render(self) -> str:
        return f"<b>{self._component.render()}</b>"


class ItalicDecorator(TextDecorator):
    """Decorator that makes text italic"""
    
    def render(self) -> str:
        return f"<i>{self._component.render()}</i>"


class UnderlineDecorator(TextDecorator):
    """Decorator that underlines text"""
    
    def render(self) -> str:
        return f"<u>{self._component.render()}</u>"


class ColorDecorator(TextDecorator):
    """Decorator that adds color to text"""
    
    def __init__(self, component: TextComponent, color: str):
        super().__init__(component)
        self.color = color
    
    def render(self) -> str:
        return f"<span style='color:{self.color}'>{self._component.render()}</span>"


# Usage
print("=== Text Formatting with Decorators ===\n")

# Plain text
text1 = PlainText("Hello World")
print(f"Plain: {text1.render()}")

# Bold text
text2 = BoldDecorator(PlainText("Hello World"))
print(f"Bold: {text2.render()}")

# Bold and Italic
text3 = ItalicDecorator(BoldDecorator(PlainText("Hello World")))
print(f"Bold + Italic: {text3.render()}")

# Bold, Italic, and Underline
text4 = UnderlineDecorator(
    ItalicDecorator(
        BoldDecorator(PlainText("Hello World"))
    )
)
print(f"Bold + Italic + Underline: {text4.render()}")

# Bold, Italic, Underline, and Red color
text5 = ColorDecorator(
    UnderlineDecorator(
        ItalicDecorator(
            BoldDecorator(PlainText("Hello World"))
        )
    ),
    "red"
)
print(f"All formats + Red: {text5.render()}")
````

## Key Benefits of Decorator Pattern

### 1. **Flexibility**
- Add or remove features dynamically at runtime
- No need to create classes for every combination

### 2. **Single Responsibility Principle**
- Each decorator has one responsibility
- Easy to understand and maintain

### 3. **Open/Closed Principle**
- Open for extension (add new decorators)
- Closed for modification (don't change existing code)

### 4. **Composition over Inheritance**
- Uses composition instead of inheritance
- More flexible than class hierarchies

### 5. **Avoids Class Explosion**
- No need for exponential number of classes
- Just combine decorators as needed


## When to Use Decorator Pattern

Use Decorator Pattern when:

1. ✅ **Need to add features dynamically**: Add responsibilities to objects at runtime
2. ✅ **Want to avoid class explosion**: Too many combinations would require too many classes
3. ✅ **Need flexible composition**: Want to mix and match features freely
4. ✅ **Can't use inheritance**: When subclassing would be impractical
5. ✅ **Want to add features without modifying existing code**: Follow Open/Closed Principle

### Common Use Cases:
- Text formatting (Bold, Italic, Underline, Color)
- Coffee/Drink add-ons (Milk, Sugar, Syrup)
- Stream processing (Buffering, Compression, Encryption)
- UI components (Scrollbars, Borders, Tooltips)
- Web middleware (Authentication, Logging, Caching)
- File I/O (Compression, Encryption, Buffering)


## Summary: Decorator Pattern

### Structure:

```
┌─────────────────────┐
│  Component (ABC)    │  ← Interface
│  + operation()      │
└─────────────────────┘
         ▲
         │
    ┌────┴────┐
    │         │
┌───┴───┐ ┌──┴──────────┐
│Concrete│ │  Decorator  │  ← Base Decorator
│Component│ │  (ABC)     │
│         │ │+ component │
└─────────┘ └────┬───────┘
                 │
          ┌──────┴──────┬──────────┬──────────┐
          │             │          │          │
    ┌─────┴─────┐ ┌────┴────┐ ┌──┴────┐ ┌──┴────┐
    │Concrete   │ │Concrete  │ │Concrete│ │Concrete│
    │Decorator  │ │Decorator │ │Decorator│ │Decorator│
    │    1      │ │    2     │ │    3    │ │    N    │
    └───────────┘ └──────────┘ └─────────┘ └─────────┘
```

### Key Components:

1. **Component Interface**: Defines the interface for objects that can have responsibilities added dynamically
2. **Concrete Component**: The base object to which decorators can be attached
3. **Decorator Base Class**: Maintains a reference to a Component and defines the decorator interface
4. **Concrete Decorators**: Add specific responsibilities to the component

### Advantages:

✅ **Flexibility**: Add features dynamically  
✅ **No Class Explosion**: Avoid exponential growth of classes  
✅ **Single Responsibility**: Each decorator has one job  
✅ **Composition**: More flexible than inheritance  
✅ **Runtime Behavior**: Change object behavior at runtime  

### Disadvantages:

❌ **Order Matters**: Decorator order can affect behavior  
❌ **Complexity**: Can be harder to understand with many decorators  
❌ **Debugging**: Harder to debug with nested decorators  
❌ **Performance**: Slight overhead due to multiple wrapper layers  

### Remember:

> **"Decorator Pattern allows you to attach new behaviors to objects by placing these objects inside special wrapper objects that contain the behaviors."**

### Python Built-in Decorators:

Python's `@decorator` syntax is actually a different concept (function decorators), but the idea is similar - wrapping functions to add functionality!

## Observer Pattern

### What is Observer Pattern?

The Observer Pattern is a behavioral design pattern that defines a one-to-many dependency between objects. When one object (subject) changes its state, all dependent objects (observers) are notified and updated automatically.

**Key Concept**: Define a dependency between objects so that when one object changes state, all its dependents are notified and updated automatically.

**Real-world analogy**: Think of a news channel (subject) and subscribers (observers). When news breaks, all subscribers are automatically notified. They can subscribe or unsubscribe at any time.


### Real-World Problem: Stock Price Notification System

Imagine you're building a stock trading application. Multiple users want to be notified when:
- Stock price changes
- Stock price reaches a certain threshold
- Stock volume changes

**Without Observer Pattern**: The stock class would need to know about all users and manually notify each one. Adding/removing users requires modifying the stock class.

**With Observer Pattern**: Users can subscribe/unsubscribe dynamically, and the stock automatically notifies all subscribers when price changes.


## Code BEFORE Observer Pattern (Problem)

Let's see how the code looks without using the Observer pattern:


````python
# ❌ BAD CODE: Without Observer Pattern

class Stock:
    """Stock class that needs to notify users"""
    
    def __init__(self, symbol, price):
        self.symbol = symbol
        self.price = price
        # Problem: Hard-coded list of users
        self.user1 = None
        self.user2 = None
        self.user3 = None
    
    def set_price(self, new_price):
        """Update price and notify all users"""
        old_price = self.price
        self.price = new_price
        
        # Problem: Must manually notify each user
        if self.user1:
            self.user1.notify(self.symbol, old_price, new_price)
        if self.user2:
            self.user2.notify(self.symbol, old_price, new_price)
        if self.user3:
            self.user3.notify(self.symbol, old_price, new_price)
    
    def add_user(self, user):
        """Add a user - but where to store it?"""
        # Problem: Limited to 3 users, what about 4th, 5th user?
        if not self.user1:
            self.user1 = user
        elif not self.user2:
            self.user2 = user
        elif not self.user3:
            self.user3 = user
        else:
            print("Cannot add more users!")


class User:
    """User class that wants to be notified"""
    
    def __init__(self, name):
        self.name = name
    
    def notify(self, symbol, old_price, new_price):
        """Receive notification"""
        print(f"{self.name} notified: {symbol} changed from ${old_price} to ${new_price}")


# Usage
print("=== Without Observer Pattern ===\n")

# Create stock
stock = Stock("AAPL", 150.0)

# Create users
user1 = User("Alice")
user2 = User("Bob")
user3 = User("Charlie")

# Add users (limited to 3!)
stock.add_user(user1)
stock.add_user(user2)
stock.add_user(user3)

# Try to add 4th user - fails!
user4 = User("David")
stock.add_user(user4)  # Cannot add!

# Update price - manually notifies each user
print("\nPrice update:")
stock.set_price(155.0)

# Problem: What if we want to remove a user? What if we want 100 users?
# This approach doesn't scale!
````

2. **Limited Scalability**: Hard-coded user limit (only 3 users)

4. **No Dynamic Subscription**: Can't easily subscribe/unsubscribe at runtime
5. **Hard to Maintain**: Adding new observer types requires changing the subject class


## Code AFTER Observer Pattern (Solution)

Now let's refactor using the Observer pattern:


````python
# ✅ GOOD CODE: With Observer Pattern

from abc import ABC, abstractmethod
from typing import List

# Step 1: Define Observer Interface
class Observer(ABC):
    """Abstract observer interface"""
    
    @abstractmethod
    def update(self, subject):
        """Update method called when subject changes"""
        pass


# Step 2: Define Subject Interface
class Subject(ABC):
    """Abstract subject interface"""
    
    @abstractmethod
    def attach(self, observer: Observer):
        """Attach an observer"""
        pass
    
    @abstractmethod
    def detach(self, observer: Observer):
        """Detach an observer"""
        pass
    
    @abstractmethod
    def notify(self):
        """Notify all observers"""
        pass


# Step 3: Create Concrete Subject
class Stock(Subject):
    """Concrete subject - stock that observers watch"""
    
    def __init__(self, symbol: str, price: float):
        self.symbol = symbol
        self._price = price
        self._observers: List[Observer] = []
    
    @property
    def price(self):
        return self._price
    
    @price.setter
    def price(self, value):
        if self._price != value:
            old_price = self._price
            self._price = value
            print(f"\n{self.symbol} price changed from ${old_price} to ${value}")
            self.notify()  # Notify all observers
    
    def attach(self, observer: Observer):
        """Subscribe an observer"""
        if observer not in self._observers:
            self._observers.append(observer)
            print(f"Observer attached: {observer.name}")
    
    def detach(self, observer: Observer):
        """Unsubscribe an observer"""
        if observer in self._observers:
            self._observers.remove(observer)
            print(f"Observer detached: {observer.name}")
    
    def notify(self):
        """Notify all observers"""
        for observer in self._observers:
            observer.update(self)


# Step 4: Create Concrete Observers
class StockObserver(Observer):
    """Concrete observer - user watching stock"""
    
    def __init__(self, name: str):
        self.name = name
    
    def update(self, subject: Stock):
        """Called when subject changes"""
        print(f"  → {self.name} notified: {subject.symbol} is now ${subject.price}")


class PriceAlertObserver(Observer):
    """Concrete observer - alerts when price threshold is reached"""
    
    def __init__(self, name: str, threshold: float, direction: str = "above"):
        self.name = name
        self.threshold = threshold
        self.direction = direction  # "above" or "below"
    
    def update(self, subject: Stock):
        """Check if price threshold is reached"""
        if self.direction == "above" and subject.price >= self.threshold:
            print(f"  → ⚠️ ALERT {self.name}: {subject.symbol} reached ${subject.price} (above ${self.threshold})")
        elif self.direction == "below" and subject.price <= self.threshold:
            print(f"  → ⚠️ ALERT {self.name}: {subject.symbol} reached ${subject.price} (below ${self.threshold})")


# Usage
print("=== Using Observer Pattern ===\n")

# Create stock
stock = Stock("AAPL", 150.0)

# Create observers
observer1 = StockObserver("Alice")
observer2 = StockObserver("Bob")
observer3 = StockObserver("Charlie")
alert_observer = PriceAlertObserver("Price Alert", 155.0, "above")

# Subscribe observers
stock.attach(observer1)
stock.attach(observer2)
stock.attach(observer3)
stock.attach(alert_observer)

# Update price - all observers notified automatically
stock.price = 152.0
stock.price = 155.0  # Triggers alert
stock.price = 158.0

# Unsubscribe an observer
print("\nUnsubscribing Bob:")
stock.detach(observer2)

# Update price again - Bob won't be notified
stock.price = 160.0
````

## More Realistic Example: News Publisher System

````python
# Example: News Publisher and Subscribers

from abc import ABC, abstractmethod
from typing import List, Dict
from datetime import datetime

class NewsObserver(ABC):
    """Observer interface for news subscribers"""
    
    @abstractmethod
    def update(self, news_item: Dict):
        """Receive news update"""
        pass


class NewsPublisher:
    """Subject - publishes news to subscribers"""
    
    def __init__(self, name: str):
        self.name = name
        self._subscribers: List[NewsObserver] = []
        self._latest_news = None
    
    def subscribe(self, observer: NewsObserver):
        """Subscribe to news"""
        if observer not in self._subscribers:
            self._subscribers.append(observer)
            print(f"✅ {observer.name} subscribed to {self.name}")
    
    def unsubscribe(self, observer: NewsObserver):
        """Unsubscribe from news"""
        if observer in self._subscribers:
            self._subscribers.remove(observer)
            print(f"❌ {observer.name} unsubscribed from {self.name}")
    
    def publish_news(self, headline: str, content: str):
        """Publish news and notify all subscribers"""
        news_item = {
            'headline': headline,
            'content': content,
            'publisher': self.name,
            'timestamp': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        }
        self._latest_news = news_item
        
        print(f"\n📰 {self.name} published: {headline}")
        print("=" * 60)
        
        # Notify all subscribers
        for subscriber in self._subscribers:
            subscriber.update(news_item)
    
    def get_subscriber_count(self) -> int:
        """Get number of subscribers"""
        return len(self._subscribers)


class EmailSubscriber(NewsObserver):
    """Concrete observer - receives news via email"""
    
    def __init__(self, name: str, email: str):
        self.name = name
        self.email = email
    
    def update(self, news_item: Dict):
        """Receive news via email"""
        print(f"  📧 Email to {self.email}: {news_item['headline']}")
        print(f"     {news_item['content'][:50]}...")


class SMSSubscriber(NewsObserver):
    """Concrete observer - receives news via SMS"""
    
    def __init__(self, name: str, phone: str):
        self.name = name
        self.phone = phone
    
    def update(self, news_item: Dict):
        """Receive news via SMS"""
        print(f"  📱 SMS to {self.phone}: {news_item['headline']}")


class PushNotificationSubscriber(NewsObserver):
    """Concrete observer - receives news via push notification"""
    
    def __init__(self, name: str, device_id: str):
        self.name = name
        self.device_id = device_id
    
    def update(self, news_item: Dict):
        """Receive news via push notification"""
        print(f"  🔔 Push to {self.device_id}: {news_item['headline']}")


# Usage
print("=== News Publisher System ===\n")

# Create publisher
tech_news = NewsPublisher("Tech News Daily")

# Create subscribers with different notification methods
alice = EmailSubscriber("Alice", "alice@example.com")
bob = SMSSubscriber("Bob", "+1234567890")
charlie = PushNotificationSubscriber("Charlie", "device_123")
diana = EmailSubscriber("Diana", "diana@example.com")

# Subscribe
tech_news.subscribe(alice)
tech_news.subscribe(bob)
tech_news.subscribe(charlie)
tech_news.subscribe(diana)

print(f"\nTotal subscribers: {tech_news.get_subscriber_count()}")

# Publish news - all subscribers notified automatically
tech_news.publish_news(
    "New AI Breakthrough",
    "Scientists have developed a new AI model that can understand context better than ever before."
)

# Unsubscribe one user
print("\n")
tech_news.unsubscribe(bob)

# Publish another news - Bob won't receive it
tech_news.publish_news(
    "Python 3.12 Released",
    "The latest version of Python includes performance improvements and new features."
)
````

## Singleton Pattern

### What is Singleton Pattern?

The Singleton Pattern is a creational design pattern that ensures a class has only one instance and provides a global point of access to that instance.

**Key Concept**: Restrict instantiation of a class to a single object. No matter how many times you try to create an instance, you always get the same one.

**Real-world analogy**: Think of a printer spooler - there should be only one spooler managing all print jobs. Or a database connection pool - you want one pool manager, not multiple competing ones.


### Real-World Problem: Database Connection Manager

Imagine you're building an application that needs to connect to a database. You want to:
- Have only ONE connection pool manager
- Share the same connection pool across the entire application
- Avoid creating multiple expensive database connections

**Without Singleton Pattern**: Every time you create a new DatabaseManager, you get a new connection pool, wasting resources and potentially causing connection limit issues.

**With Singleton Pattern**: No matter where you create DatabaseManager, you always get the same instance with the same connection pool.


## Code BEFORE Singleton Pattern (Problem)

Let's see how the code looks without using the Singleton pattern:


````python
# ❌ BAD CODE: Without Singleton Pattern

class DatabaseConnection:
    """Database connection class - creates new instance every time"""
    
    def __init__(self):
        print("Creating new database connection...")
        print("  - Establishing connection to database...")
        print("  - Creating connection pool...")
        print("  - Initializing connection manager...")
        self.connection_string = "postgresql://localhost:5432/mydb"
        self.pool_size = 10
        self.connections = []
        print("  ✅ Database connection created!")
    
    def connect(self):
        """Connect to database"""
        print(f"Using connection: {self.connection_string}")
        return f"Connected to {self.connection_string}"
    
    def get_pool_size(self):
        """Get connection pool size"""
        return self.pool_size


# Usage
print("=== Without Singleton Pattern ===\n")

# Problem: Each time we create DatabaseConnection, we get a NEW instance
db1 = DatabaseConnection()
print(f"db1 id: {id(db1)}")
print()

db2 = DatabaseConnection()
print(f"db2 id: {id(db2)}")
print()

db3 = DatabaseConnection()
print(f"db3 id: {id(db3)}")
print()

# Problem: These are DIFFERENT objects!
print(f"db1 == db2? {db1 is db2}")  # False - different instances!
print(f"db2 == db3? {db2 is db3}")  # False - different instances!

# Problem: We're creating multiple expensive database connections!
# This wastes resources and can cause connection limit issues!
````

### Problems with the Above Code:

1. **Resource Waste**: Creates multiple expensive connections unnecessarily
2. **Connection Limit Issues**: Database may have connection limits
3. **Inconsistent State**: Each instance has its own state
4. **Memory Overhead**: Multiple instances consume more memory
5. **No Global Access**: Can't ensure same instance is used everywhere


## Code AFTER Singleton Pattern (Solution)

Now let's refactor using the Singleton pattern:


````python
# ✅ GOOD CODE: With Singleton Pattern

class DatabaseConnection:
    """Singleton Database Connection"""
    
    _instance = None  # Class variable to store the single instance
    
    def __new__(cls):
        """Override __new__ to control instance creation"""
        if cls._instance is None:
            print("Creating new database connection (first time only)...")
            cls._instance = super(DatabaseConnection, cls).__new__(cls)
            cls._instance._initialized = False
        return cls._instance
    
    def __init__(self):
        """Initialize only once"""
        if not self._initialized:
            print("  - Establishing connection to database...")
            print("  - Creating connection pool...")
            print("  - Initializing connection manager...")
            self.connection_string = "postgresql://localhost:5432/mydb"
            self.pool_size = 10
            self.connections = []
            self._initialized = True
            print("  ✅ Database connection created!")
        else:
            print("  ℹ️ Using existing database connection...")
    
    def connect(self):
        """Connect to database"""
        print(f"Using connection: {self.connection_string}")
        return f"Connected to {self.connection_string}"
    
    def get_pool_size(self):
        """Get connection pool size"""
        return self.pool_size


# Usage
print("=== Using Singleton Pattern ===\n")

# First creation - creates new instance
db1 = DatabaseConnection()
print(f"db1 id: {id(db1)}")
print()

# Second creation - returns SAME instance
db2 = DatabaseConnection()
print(f"db2 id: {id(db2)}")
print()

# Third creation - returns SAME instance
db3 = DatabaseConnection()
print(f"db3 id: {id(db3)}")
print()

# All are the SAME object!
print(f"db1 == db2? {db1 is db2}")  # True - same instance!
print(f"db2 == db3? {db2 is db3}")  # True - same instance!
print(f"All same? {db1 is db2 is db3}")  # True!

# Modify one, all reflect the change
db1.pool_size = 20
print(f"\ndb1.pool_size = {db1.pool_size}")
print(f"db2.pool_size = {db2.pool_size}")  # Same value!
print(f"db3.pool_size = {db3.pool_size}")  # Same value!
````

## Alternative Implementation: Using Decorator

Python provides a cleaner way using a decorator:


````python
# Alternative: Singleton using Decorator (Pythonic way)

def singleton(cls):
    """Decorator to make a class singleton"""
    instances = {}
    
    def get_instance(*args, **kwargs):
        if cls not in instances:
            instances[cls] = cls(*args, **kwargs)
        return instances[cls]
    
    return get_instance


@singleton
class Logger:
    """Singleton Logger class"""
    
    def __init__(self):
        print("Initializing Logger (only once)...")
        self.log_file = "app.log"
        self.log_level = "INFO"
        self.logs = []
    
    def log(self, message: str, level: str = "INFO"):
        """Log a message"""
        log_entry = f"[{level}] {message}"
        self.logs.append(log_entry)
        print(f"Logging: {log_entry}")
    
    def get_logs(self):
        """Get all logs"""
        return self.logs


# Usage
print("=== Singleton using Decorator ===\n")

# All these create the SAME instance
logger1 = Logger()
logger2 = Logger()
logger3 = Logger()

print(f"\nlogger1 is logger2? {logger1 is logger2}")  # True
print(f"logger2 is logger3? {logger2 is logger3}")  # True

# All share the same logs
logger1.log("First log message")
logger2.log("Second log message")
logger3.log("Third log message")

print(f"\nAll logs from logger1: {logger1.get_logs()}")
print(f"All logs from logger2: {logger2.get_logs()}")  # Same logs!
````

````python
# Example: Configuration Manager Singleton

class ConfigManager:
    """Singleton Configuration Manager"""
    
    _instance = None
    _initialized = False
    
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super(ConfigManager, cls).__new__(cls)
        return cls._instance
    
    def __init__(self):
        if not ConfigManager._initialized:
            print("🔧 Initializing Configuration Manager...")
            self.config = {
                'database_url': 'postgresql://localhost:5432/mydb',
                'api_key': 'secret_key_123',
                'debug_mode': False,
                'max_connections': 100,
                'timeout': 30
            }
            ConfigManager._initialized = True
            print("✅ Configuration loaded!")
        else:
            print("ℹ️ Using existing configuration...")
    
    def get(self, key: str):
        """Get configuration value"""
        return self.config.get(key)
    
    def set(self, key: str, value):
        """Set configuration value"""
        self.config[key] = value
        print(f"⚙️ Updated {key} = {value}")
    
    def get_all(self):
        """Get all configuration"""
        return self.config.copy()


# Usage
print("=== Configuration Manager Singleton ===\n")

# Create config managers from different parts of application
config1 = ConfigManager()  # First time - initializes
print()

config2 = ConfigManager()  # Second time - uses existing
print()

config3 = ConfigManager()  # Third time - uses existing
print()

# All are the same instance
print(f"config1 is config2? {config1 is config2}")
print(f"config2 is config3? {config2 is config3}")

# Access configuration from anywhere
print(f"\nDatabase URL: {config1.get('database_url')}")
print(f"API Key: {config2.get('api_key')}")
print(f"Max Connections: {config3.get('max_connections')}")

# Update from one instance, all see the change
print("\nUpdating configuration via config1:")
config1.set('debug_mode', True)
print(f"config2 sees debug_mode: {config2.get('debug_mode')}")
print(f"config3 sees debug_mode: {config3.get('debug_mode')}")

# All instances share the same configuration
print(f"\nAll configs are the same: {config1.get_all() == config2.get_all() == config3.get_all()}")
````

## Thread-Safe Singleton (For Multi-threading)

In multi-threaded applications, you need thread-safe singleton:


````python
# Thread-Safe Singleton Implementation

import threading

class ThreadSafeSingleton:
    """Thread-safe Singleton using lock"""
    
    _instance = None
    _lock = threading.Lock()
    _initialized = False
    
    def __new__(cls):
        if cls._instance is None:
            with cls._lock:
                # Double-check locking pattern
                if cls._instance is None:
                    cls._instance = super(ThreadSafeSingleton, cls).__new__(cls)
        return cls._instance
    
    def __init__(self):
        if not ThreadSafeSingleton._initialized:
            with ThreadSafeSingleton._lock:
                if not ThreadSafeSingleton._initialized:
                    print("Initializing ThreadSafeSingleton...")
                    self.data = []
                    ThreadSafeSingleton._initialized = True
    
    def add_data(self, item):
        """Add data (thread-safe)"""
        with self._lock:
            self.data.append(item)
            print(f"Added: {item}")
    
    def get_data(self):
        """Get all data"""
        return self.data.copy()


# Usage in multi-threaded environment
print("=== Thread-Safe Singleton ===\n")

def create_singleton(thread_id):
    """Function to create singleton from different threads"""
    instance = ThreadSafeSingleton()
    instance.add_data(f"Data from Thread {thread_id}")
    return instance

# Simulate multiple threads
threads = []
instances = []

for i in range(5):
    thread = threading.Thread(target=lambda i=i: instances.append(create_singleton(i)))
    threads.append(thread)
    thread.start()

# Wait for all threads
for thread in threads:
    thread.join()

# All instances should be the same
print(f"\nAll instances same? {all(instances[0] is inst for inst in instances)}")
print(f"Total data items: {len(instances[0].get_data())}")
````

         │

### Advantages:

### Remember:

### Python Best Practices:

## Facade Pattern

### What is Facade Pattern?

The Facade Pattern is a structural design pattern that provides a simplified interface to a complex subsystem. It hides the complexity of the subsystem and provides a unified interface to access its functionality.

**Key Concept**: Provide a simple, easy-to-use interface over a large and sophisticated body of code. It acts as a "front door" to a complex subsystem.

**Real-world analogy**: Think of a home theater system - you have multiple components (TV, DVD player, sound system, lights). Instead of controlling each separately, a remote control (facade) provides a simple "Watch Movie" button that handles all the complexity.


### Real-World Problem: Home Theater System

Imagine you're building a home theater system. To watch a movie, you need to:
- Turn on the TV
- Turn on the DVD player
- Set the sound system to surround sound
- Dim the lights
- Start the movie

**Without Facade Pattern**: Users must interact with each component separately, knowing all the details and order of operations.

**With Facade Pattern**: Users just call "watchMovie()" and the facade handles all the complexity behind the scenes.


## Code BEFORE Facade Pattern (Problem)

Let's see how the code looks without using the Facade pattern:

## More Realistic Example: Order Processing System

Let's see another practical example with an e-commerce order processing system:


````python
# Example: E-commerce Order Processing with Facade

class InventoryService:
    """Subsystem: Inventory management"""
    def check_availability(self, product_id, quantity):
        print(f"📦 Checking inventory for product {product_id}...")
        # Simulate check
        available = True
        if available:
            print(f"   ✅ {quantity} units available")
            return True
        return False
    
    def reserve_items(self, product_id, quantity):
        print(f"   🔒 Reserving {quantity} units of product {product_id}")
        return True
    
    def release_reservation(self, product_id, quantity):
        print(f"   🔓 Releasing reservation for {quantity} units")


class PaymentService:
    """Subsystem: Payment processing"""
    def validate_payment(self, card_number, amount):
        print(f"💳 Validating payment...")
        print(f"   Card: ****{card_number[-4:]}")
        print(f"   Amount: ${amount}")
        # Simulate validation
        if len(card_number) == 16:
            print("   ✅ Payment validated")
            return True
        return False
    
    def process_payment(self, card_number, amount):
        print(f"   💰 Processing payment of ${amount}...")
        print("   ✅ Payment processed successfully")
        return True
    
    def refund_payment(self, transaction_id, amount):
        print(f"   💸 Refunding ${amount} for transaction {transaction_id}")


class ShippingService:
    """Subsystem: Shipping management"""
    def calculate_shipping(self, address, weight):
        print(f"🚚 Calculating shipping...")
        print(f"   Address: {address}")
        print(f"   Weight: {weight} kg")
        shipping_cost = weight * 5  # $5 per kg
        print(f"   💵 Shipping cost: ${shipping_cost}")
        return shipping_cost
    
    def create_shipment(self, order_id, address):
        print(f"   📦 Creating shipment for order {order_id}")
        print(f"   📍 Shipping to: {address}")
        tracking_number = f"TRACK{order_id}"
        print(f"   🏷️ Tracking number: {tracking_number}")
        return tracking_number


class NotificationService:
    """Subsystem: Notifications"""
    def send_email(self, email, subject, message):
        print(f"📧 Sending email to {email}")
        print(f"   Subject: {subject}")
        print(f"   Message: {message[:50]}...")
    
    def send_sms(self, phone, message):
        print(f"📱 Sending SMS to {phone}")
        print(f"   Message: {message[:50]}...")


# Facade: Simplifies order processing
class OrderFacade:
    """Facade that simplifies order processing"""
    
    def __init__(self):
        self.inventory = InventoryService()
        self.payment = PaymentService()
        self.shipping = ShippingService()
        self.notification = NotificationService()
    
    def place_order(self, product_id, quantity, card_number, address, customer_email, customer_phone):
        """Simple method to place an order - handles all complexity"""
        print("=" * 60)
        print("🛒 Processing Order")
        print("=" * 60 + "\n")
        
        order_id = f"ORD{hash(product_id) % 10000}"
        total_amount = quantity * 100  # $100 per item
        
        # Step 1: Check inventory
        if not self.inventory.check_availability(product_id, quantity):
            print("❌ Order failed: Insufficient inventory")
            return False
        
        # Step 2: Reserve items
        self.inventory.reserve_items(product_id, quantity)
        
        # Step 3: Calculate shipping
        shipping_cost = self.shipping.calculate_shipping(address, quantity * 0.5)
        total_amount += shipping_cost
        
        # Step 4: Validate payment
        if not self.payment.validate_payment(card_number, total_amount):
            print("❌ Order failed: Payment validation failed")
            self.inventory.release_reservation(product_id, quantity)
            return False
        
        # Step 5: Process payment
        if not self.payment.process_payment(card_number, total_amount):
            print("❌ Order failed: Payment processing failed")
            self.inventory.release_reservation(product_id, quantity)
            return False
        
        # Step 6: Create shipment
        tracking_number = self.shipping.create_shipment(order_id, address)
        
        # Step 7: Send notifications
        self.notification.send_email(
            customer_email,
            "Order Confirmation",
            f"Your order {order_id} has been confirmed. Tracking: {tracking_number}"
        )
        self.notification.send_sms(
            customer_phone,
            f"Order {order_id} confirmed. Track: {tracking_number}"
        )
        
        print("\n" + "=" * 60)
        print(f"✅ Order {order_id} placed successfully!")
        print(f"   Total: ${total_amount}")
        print(f"   Tracking: {tracking_number}")
        print("=" * 60 + "\n")
        
        return order_id


# Usage
print("=== E-commerce Order Processing ===\n")

# Create facade
order_system = OrderFacade()

# Client just calls one simple method - facade handles everything!
order_system.place_order(
    product_id="PROD123",
    quantity=2,
    card_number="1234567890123456",
    address="123 Main St, City, Country",
    customer_email="customer@example.com",
    customer_phone="+1234567890"
)
````

### Remember:

## Factory Pattern

### What is Factory Pattern?

The Factory Pattern is a creational design pattern that provides an interface for creating objects without specifying their exact classes. It encapsulates object creation logic and lets subclasses decide which class to instantiate.

**Key Concept**: Instead of using `new` operator directly, use a factory method to create objects. This allows you to create objects without exposing the instantiation logic to the client.

**Real-world analogy**: Think of a pizza restaurant - you don't go into the kitchen to make your own pizza. You order from the menu, and the kitchen (factory) creates the pizza for you based on your order type.


### Real-World Problem: Vehicle Manufacturing

Imagine you're building a vehicle manufacturing system. You need to create different types of vehicles:
- Car
- Motorcycle
- Truck
- Bus

**Without Factory Pattern**: Client code must know about all vehicle classes and use conditional logic to create the right type. Adding a new vehicle type requires modifying client code.

**With Factory Pattern**: Client just requests a vehicle type, and the factory creates the appropriate vehicle. Adding new vehicle types only requires updating the factory.


## Code BEFORE Factory Pattern (Problem)

Let's see how the code looks without using the Factory pattern:


````python
# ❌ BAD CODE: Without Factory Pattern

class Car:
    """Car class"""
    def __init__(self):
        self.type = "Car"
        self.wheels = 4
    
    def drive(self):
        print(f"{self.type} is driving with {self.wheels} wheels")


class Motorcycle:
    """Motorcycle class"""
    def __init__(self):
        self.type = "Motorcycle"
        self.wheels = 2
    
    def drive(self):
        print(f"{self.type} is driving with {self.wheels} wheels")


class Truck:
    """Truck class"""
    def __init__(self):
        self.type = "Truck"
        self.wheels = 6
    
    def drive(self):
        print(f"{self.type} is driving with {self.wheels} wheels")


# Usage - Client must know about all classes
print("=== Without Factory Pattern ===\n")

# Problem: Client code has conditional logic and knows about all classes
def create_vehicle(vehicle_type):
    """Create vehicle - but client knows about all classes!"""
    if vehicle_type == "car":
        return Car()
    elif vehicle_type == "motorcycle":
        return Motorcycle()
    elif vehicle_type == "truck":
        return Truck()
    else:
        raise ValueError(f"Unknown vehicle type: {vehicle_type}")

# Create vehicles
vehicle1 = create_vehicle("car")
vehicle1.drive()

vehicle2 = create_vehicle("motorcycle")
vehicle2.drive()

vehicle3 = create_vehicle("truck")
vehicle3.drive()

# Problem: What if we want to add a new vehicle type (Bus)?
# We must modify the create_vehicle function!
# This violates Open/Closed Principle!
````

### Problems with the Above Code:

1. **Tight Coupling**: Client code is tightly coupled to concrete vehicle classes
2. **Violates Open/Closed Principle**: Must modify client code to add new vehicle types
3. **Complex Conditional Logic**: If-else chains become hard to maintain
4. **Hard to Test**: Difficult to mock or test different vehicle types
5. **Code Duplication**: Creation logic scattered across codebase


## Code AFTER Factory Pattern (Solution)

Now let's refactor using the Factory pattern:


````python
# ✅ GOOD CODE: With Factory Pattern

from abc import ABC, abstractmethod

# Step 1: Define Product Interface
class Vehicle(ABC):
    """Abstract product interface"""
    
    @abstractmethod
    def drive(self):
        """Drive the vehicle"""
        pass
    
    @abstractmethod
    def get_info(self):
        """Get vehicle information"""
        pass


# Step 2: Create Concrete Products
class Car(Vehicle):
    """Concrete product - Car"""
    
    def __init__(self):
        self.type = "Car"
        self.wheels = 4
        self.max_speed = 120
    
    def drive(self):
        print(f"🚗 {self.type} is driving with {self.wheels} wheels")
    
    def get_info(self):
        return f"{self.type}: {self.wheels} wheels, max speed {self.max_speed} km/h"


class Motorcycle(Vehicle):
    """Concrete product - Motorcycle"""
    
    def __init__(self):
        self.type = "Motorcycle"
        self.wheels = 2
        self.max_speed = 150
    
    def drive(self):
        print(f"🏍️ {self.type} is driving with {self.wheels} wheels")
    
    def get_info(self):
        return f"{self.type}: {self.wheels} wheels, max speed {self.max_speed} km/h"


class Truck(Vehicle):
    """Concrete product - Truck"""
    
    def __init__(self):
        self.type = "Truck"
        self.wheels = 6
        self.max_speed = 80
    
    def drive(self):
        print(f"🚚 {self.type} is driving with {self.wheels} wheels")
    
    def get_info(self):
        return f"{self.type}: {self.wheels} wheels, max speed {self.max_speed} km/h"


# Step 3: Create Factory
class VehicleFactory:
    """Factory class that creates vehicles"""
    
    @staticmethod
    def create_vehicle(vehicle_type: str) -> Vehicle:
        """Factory method to create vehicles"""
        vehicles = {
            "car": Car,
            "motorcycle": Motorcycle,
            "truck": Truck
        }
        
        vehicle_class = vehicles.get(vehicle_type.lower())
        if vehicle_class:
            return vehicle_class()
        else:
            raise ValueError(f"Unknown vehicle type: {vehicle_type}")


# Usage - Much cleaner!
print("=== Using Factory Pattern ===\n")

# Client just uses factory - doesn't know about concrete classes
factory = VehicleFactory()

vehicle1 = factory.create_vehicle("car")
print(vehicle1.get_info())
vehicle1.drive()
print()

vehicle2 = factory.create_vehicle("motorcycle")
print(vehicle2.get_info())
vehicle2.drive()
print()

vehicle3 = factory.create_vehicle("truck")
print(vehicle3.get_info())
vehicle3.drive()

# To add a new vehicle type (Bus), just:
# 1. Create Bus class inheriting from Vehicle
# 2. Add it to the factory's vehicles dictionary
# No need to modify client code!
````

## Factory Method Pattern (More Advanced)

The Factory Method Pattern uses inheritance - each factory subclass creates a specific product:


````python
# Factory Method Pattern (Using Inheritance)

from abc import ABC, abstractmethod

# Product Interface
class Document(ABC):
    """Abstract document interface"""
    
    @abstractmethod
    def open(self):
        """Open document"""
        pass
    
    @abstractmethod
    def save(self):
        """Save document"""
        pass


# Concrete Products
class PDFDocument(Document):
    """PDF document"""
    
    def open(self):
        print("📄 Opening PDF document...")
    
    def save(self):
        print("💾 Saving PDF document...")


class WordDocument(Document):
    """Word document"""
    
    def open(self):
        print("📝 Opening Word document...")
    
    def save(self):
        print("💾 Saving Word document...")


class ExcelDocument(Document):
    """Excel document"""
    
    def open(self):
        print("📊 Opening Excel document...")
    
    def save(self):
        print("💾 Saving Excel document...")


# Abstract Factory
class DocumentFactory(ABC):
    """Abstract factory interface"""
    
    @abstractmethod
    def create_document(self) -> Document:
        """Factory method to create document"""
        pass


# Concrete Factories
class PDFFactory(DocumentFactory):
    """Factory for creating PDF documents"""
    
    def create_document(self) -> Document:
        return PDFDocument()


class WordFactory(DocumentFactory):
    """Factory for creating Word documents"""
    
    def create_document(self) -> Document:
        return WordDocument()


class ExcelFactory(DocumentFactory):
    """Factory for creating Excel documents"""
    
    def create_document(self) -> Document:
        return ExcelDocument()


# Usage
print("=== Factory Method Pattern ===\n")

# Create factories
pdf_factory = PDFFactory()
word_factory = WordFactory()
excel_factory = ExcelFactory()

# Use factories to create documents
pdf = pdf_factory.create_document()
pdf.open()
pdf.save()
print()

word = word_factory.create_document()
word.open()
word.save()
print()

excel = excel_factory.create_document()
excel.open()
excel.save()
````

## More Realistic Example: Payment Gateway Factory

Let's see a practical example with payment gateway integration:


````python
# Example: Payment Gateway Factory

from abc import ABC, abstractmethod

# Product Interface
class PaymentGateway(ABC):
    """Abstract payment gateway interface"""
    
    @abstractmethod
    def process_payment(self, amount: float, currency: str) -> bool:
        """Process payment"""
        pass
    
    @abstractmethod
    def refund_payment(self, transaction_id: str, amount: float) -> bool:
        """Refund payment"""
        pass


# Concrete Products
class StripeGateway(PaymentGateway):
    """Stripe payment gateway"""
    
    def __init__(self, api_key: str):
        self.api_key = api_key
        self.name = "Stripe"
    
    def process_payment(self, amount: float, currency: str) -> bool:
        print(f"💳 Processing ${amount} {currency} via Stripe...")
        print(f"   Using API key: {self.api_key[:10]}...")
        print("   ✅ Payment processed successfully")
        return True
    
    def refund_payment(self, transaction_id: str, amount: float) -> bool:
        print(f"💸 Refunding ${amount} via Stripe for transaction {transaction_id}")
        print("   ✅ Refund processed successfully")
        return True


class PayPalGateway(PaymentGateway):
    """PayPal payment gateway"""
    
    def __init__(self, client_id: str, client_secret: str):
        self.client_id = client_id
        self.client_secret = client_secret
        self.name = "PayPal"
    
    def process_payment(self, amount: float, currency: str) -> bool:
        print(f"💳 Processing ${amount} {currency} via PayPal...")
        print(f"   Using Client ID: {self.client_id[:10]}...")
        print("   ✅ Payment processed successfully")
        return True
    
    def refund_payment(self, transaction_id: str, amount: float) -> bool:
        print(f"💸 Refunding ${amount} via PayPal for transaction {transaction_id}")
        print("   ✅ Refund processed successfully")
        return True


class SquareGateway(PaymentGateway):
    """Square payment gateway"""
    
    def __init__(self, access_token: str):
        self.access_token = access_token
        self.name = "Square"
    
    def process_payment(self, amount: float, currency: str) -> bool:
        print(f"💳 Processing ${amount} {currency} via Square...")
        print(f"   Using Access Token: {self.access_token[:10]}...")
        print("   ✅ Payment processed successfully")
        return True
    
    def refund_payment(self, transaction_id: str, amount: float) -> bool:
        print(f"💸 Refunding ${amount} via Square for transaction {transaction_id}")
        print("   ✅ Refund processed successfully")
        return True


# Factory
class PaymentGatewayFactory:
    """Factory for creating payment gateways"""
    
    @staticmethod
    def create_gateway(gateway_type: str, **credentials) -> PaymentGateway:
        """Factory method to create payment gateway"""
        
        gateways = {
            "stripe": lambda: StripeGateway(credentials.get("api_key", "")),
            "paypal": lambda: PayPalGateway(
                credentials.get("client_id", ""),
                credentials.get("client_secret", "")
            ),
            "square": lambda: SquareGateway(credentials.get("access_token", ""))
        }
        
        gateway_creator = gateways.get(gateway_type.lower())
        if gateway_creator:
            return gateway_creator()
        else:
            raise ValueError(f"Unknown payment gateway: {gateway_type}")


# Usage
print("=== Payment Gateway Factory ===\n")

# Create different payment gateways using factory
stripe = PaymentGatewayFactory.create_gateway(
    "stripe",
    api_key="sk_test_1234567890"
)
stripe.process_payment(100.0, "USD")
print()

paypal = PaymentGatewayFactory.create_gateway(
    "paypal",
    client_id="paypal_client_123",
    client_secret="paypal_secret_456"
)
paypal.process_payment(200.0, "USD")
print()

square = PaymentGatewayFactory.create_gateway(
    "square",
    access_token="sq_access_token_789"
)
square.process_payment(150.0, "USD")

# Client code doesn't need to know about concrete gateway classes!
# Easy to add new payment gateways without changing client code!
````

### Structure:

```
┌─────────────────────┐
│   Product (ABC)     │  ← Interface
│  + operation()      │
└─────────────────────┘
         ▲
         │ implements
         │
    ┌────┴────┬──────────┬──────────┐
    │         │          │          │
┌───┴───┐ ┌──┴───┐  ┌───┴───┐  ┌───┴───┐
│Product│ │Product│ │Product│ │Product│
│   1   │ │   2   │ │   3   │ │   N   │
└───────┘ └───────┘ └───────┘ └───────┘
         ▲
         │ created by
         │
┌────────┴────────┐
│   Factory        │
│  + create()      │  ← Factory method
└──────────────────┘
```

### Key Components:

1. **Product Interface**: Defines the interface for objects created by factory
2. **Concrete Products**: Classes that implement the product interface
3. **Factory**: Class with method to create products
4. **Client**: Uses factory to create products without knowing concrete classes

### Advantages:

✅ **Loose Coupling**: Client doesn't depend on concrete classes  
✅ **Single Responsibility**: Creation logic centralized  
✅ **Open/Closed Principle**: Easy to extend with new products  
✅ **Flexibility**: Can change creation logic easily  
✅ **Code Reusability**: Creation logic is reusable  
✅ **Centralized Control**: All object creation in one place  

### Disadvantages:

❌ **Additional Complexity**: Adds another layer of abstraction  
❌ **More Classes**: Need to create factory classes  
❌ **Overkill for Simple Cases**: May be unnecessary for simple object creation  

### Remember:

> **"Factory Pattern provides an interface for creating objects without specifying their exact classes. It encapsulates object creation logic."**

### Factory Pattern Variants:

1. **Simple Factory**: One factory class, one method
2. **Factory Method**: Abstract factory with subclasses
3. **Abstract Factory**: Creates families of related objects
4. **Static Factory**: Factory methods are static

### Python Best Practices:

- Use `@staticmethod` for simple factory methods
- Use abstract base classes for product interfaces
- Consider using `enum` for product types
- Use dictionary mapping for simple factories
- Consider dependency injection for complex scenarios

## Adapter Pattern

### What is Adapter Pattern?

The Adapter Pattern is a structural design pattern that allows objects with incompatible interfaces to work together. It acts as a bridge between two incompatible interfaces by wrapping an object with an adapter that makes it compatible with another class.

**Key Concept**: Convert the interface of a class into another interface that clients expect. Adapter lets classes work together that couldn't otherwise because of incompatible interfaces.

**Real-world analogy**: Think of a power adapter - your phone charger has a USB plug, but the wall outlet has a different socket. An adapter converts the outlet's interface to match your charger's interface, allowing them to work together.


### Real-World Problem: Media Player System

Imagine you have a media player that can only play MP3 files, but you have:
- MP4 files
- VLC files
- AVI files

**Without Adapter Pattern**: You'd need to modify the media player to support each format, or create separate players for each format.

**With Adapter Pattern**: You create adapters that convert each format to MP3-compatible interface, allowing the existing player to work with all formats.


## Code BEFORE Adapter Pattern (Problem)

Let's see how the code looks without using the Adapter pattern:


````python
# ❌ BAD CODE: Without Adapter Pattern

class MP3Player:
    """Media player that only plays MP3 files"""
    
    def play_mp3(self, filename):
        """Play MP3 file"""
        print(f"🎵 Playing MP3 file: {filename}")


class MP4Player:
    """MP4 player with different interface"""
    
    def play_mp4(self, filename):
        """Play MP4 file"""
        print(f"🎬 Playing MP4 file: {filename}")


class VLCPlayer:
    """VLC player with different interface"""
    
    def play_vlc(self, filename):
        """Play VLC file"""
        print(f"📹 Playing VLC file: {filename}")


# Usage - Problem: Incompatible interfaces!
print("=== Without Adapter Pattern ===\n")

mp3_player = MP3Player()
mp4_player = MP4Player()
vlc_player = VLCPlayer()

# Each player has different method names
mp3_player.play_mp3("song.mp3")
mp4_player.play_mp4("movie.mp4")
vlc_player.play_vlc("video.vlc")

# Problem: If you want to use a single interface to play all formats,
# you need to know which player to use and which method to call!
# This creates tight coupling and makes code complex!
````

### Problems with the Above Code:

1. **Incompatible Interfaces**: Different players have different method names
2. **Tight Coupling**: Client code must know about all player types
3. **Complex Logic**: Need conditional logic to choose the right player
4. **Hard to Extend**: Adding new formats requires modifying client code
5. **Violates Open/Closed Principle**: Must change existing code to support new formats


## Code AFTER Adapter Pattern (Solution)

Now let's refactor using the Adapter pattern:


````python
# ✅ GOOD CODE: With Adapter Pattern

# Step 1: Define Target Interface (what client expects)
class MediaPlayer:
    """Target interface - what client expects"""
    
    def play(self, audio_type: str, filename: str):
        """Play media file"""
        pass


# Step 2: Existing classes with incompatible interfaces
class MP3Player:
    """Adaptee - existing class with incompatible interface"""
    
    def play_mp3(self, filename):
        """Play MP3 file"""
        print(f"🎵 Playing MP3 file: {filename}")


class MP4Player:
    """Adaptee - existing class with incompatible interface"""
    
    def play_mp4(self, filename):
        """Play MP4 file"""
        print(f"🎬 Playing MP4 file: {filename}")


class VLCPlayer:
    """Adaptee - existing class with incompatible interface"""
    
    def play_vlc(self, filename):
        """Play VLC file"""
        print(f"📹 Playing VLC file: {filename}")


# Step 3: Create Adapters
class MP3Adapter(MediaPlayer):
    """Adapter for MP3 player"""
    
    def __init__(self):
        self.mp3_player = MP3Player()
    
    def play(self, audio_type: str, filename: str):
        if audio_type == "mp3":
            self.mp3_player.play_mp3(filename)
        else:
            print(f"❌ MP3 adapter cannot play {audio_type} format")


class MP4Adapter(MediaPlayer):
    """Adapter for MP4 player"""
    
    def __init__(self):
        self.mp4_player = MP4Player()
    
    def play(self, audio_type: str, filename: str):
        if audio_type == "mp4":
            self.mp4_player.play_mp4(filename)
        else:
            print(f"❌ MP4 adapter cannot play {audio_type} format")


class VLCAdapter(MediaPlayer):
    """Adapter for VLC player"""
    
    def __init__(self):
        self.vlc_player = VLCPlayer()
    
    def play(self, audio_type: str, filename: str):
        if audio_type == "vlc":
            self.vlc_player.play_vlc(filename)
        else:
            print(f"❌ VLC adapter cannot play {audio_type} format")


# Step 4: Client code uses unified interface
class MediaPlayerClient:
    """Client that uses unified interface"""
    
    def __init__(self):
        self.adapters = {
            "mp3": MP3Adapter(),
            "mp4": MP4Adapter(),
            "vlc": VLCAdapter()
        }
    
    def play_media(self, audio_type: str, filename: str):
        """Play media using unified interface"""
        adapter = self.adapters.get(audio_type.lower())
        if adapter:
            adapter.play(audio_type.lower(), filename)
        else:
            print(f"❌ Unsupported format: {audio_type}")


# Usage - Much cleaner!
print("=== Using Adapter Pattern ===\n")

player = MediaPlayerClient()

# All use the same interface!
player.play_media("mp3", "song.mp3")
player.play_media("mp4", "movie.mp4")
player.play_media("vlc", "video.vlc")

# Client doesn't need to know about different player classes!
# Just uses the unified interface!
````

## Object Adapter vs Class Adapter

There are two types of adapters:

1. **Object Adapter** (Composition): Uses composition - adapter contains an instance of adaptee
2. **Class Adapter** (Inheritance): Uses inheritance - adapter inherits from both target and adaptee

Python typically uses Object Adapter (composition) since it doesn't support multiple inheritance well.

## Summary: Observer Pattern

### Structure:

```
┌─────────────────────┐
│   Subject (ABC)     │  ← Interface
│  + attach()         │
│  + detach()         │
│  + notify()         │
└─────────────────────┘
         ▲
         │ implements
         │
┌────────┴────────┐
│ Concrete Subject │
│  + state         │
│  + observers[]   │
└─────────────────┘
         │
         │ notifies
         ▼
┌─────────────────────┐
│   Observer (ABC)     │  ← Interface
│  + update()          │
└─────────────────────┘
         ▲
         │ implements
         │
    ┌────┴────┬──────────┬──────────┐
    │         │          │          │
┌───┴───┐ ┌──┴───┐  ┌───┴───┐  ┌───┴───┐
│Concrete│ │Concrete│ │Concrete│ │Concrete│
│Observer│ │Observer│ │Observer│ │Observer│
│   1    │ │   2    │ │   3    │ │   N    │
└────────┘ └────────┘ └────────┘ └────────┘
```

### Key Components:

1. **Subject Interface**: Defines methods to attach, detach, and notify observers
2. **Concrete Subject**: Maintains list of observers and notifies them of state changes
3. **Observer Interface**: Defines update method that observers must implement
4. **Concrete Observers**: Implement update method to react to subject's state changes

### Advantages:

✅ **Loose Coupling**: Subject and observers are loosely coupled  
✅ **Dynamic Relationships**: Add/remove observers at runtime  
✅ **Broadcast Communication**: One-to-many dependency automatically maintained  
✅ **Open/Closed Principle**: Easy to add new observers without modifying subject  
✅ **Separation of Concerns**: Subject and observers have separate responsibilities  

### Disadvantages:

❌ **Unexpected Updates**: Observers may receive updates they don't care about  
❌ **Performance**: Notifying many observers can be slow  
❌ **Memory Leaks**: Forgetting to detach observers can cause memory leaks  
❌ **Update Order**: No guarantee of update order among observers  
❌ **Circular Dependencies**: Can create circular dependencies if not careful  

### Remember:

> **"Observer Pattern defines a one-to-many dependency between objects so that when one object changes state, all its dependents are notified and updated automatically."**

### Python Built-in Support:

Python's `asyncio` and event-driven frameworks use observer-like patterns. Also, many GUI frameworks (Tkinter, PyQt) use observer pattern for event handling!


## Key Benefits of Adapter Pattern

### 1. **Compatibility**
- Makes incompatible interfaces work together
- Allows integration of third-party libraries

### 2. **Reusability**
- Reuse existing classes with incompatible interfaces
- Don't need to modify existing code

### 3. **Single Responsibility**
- Adapter has one job: convert interface
- Separates interface conversion from business logic

### 4. **Open/Closed Principle**
- Open for extension (add new adapters)
- Closed for modification (don't change existing code)

### 5. **Decoupling**
- Client code doesn't depend on adaptee classes
- Client only depends on target interface


## When to Use Adapter Pattern

Use Adapter Pattern when:

1. ✅ **Incompatible interfaces**: Need to use a class with incompatible interface
2. ✅ **Third-party integration**: Integrating third-party libraries with different interfaces
3. ✅ **Legacy code**: Need to use legacy code with modern interfaces
4. ✅ **Multiple formats**: Need to support multiple formats with unified interface
5. ✅ **Reuse existing code**: Want to reuse existing classes without modifying them

### Common Use Cases:
- Integrating payment gateways (Stripe, PayPal, Square)
- Database adapters (MySQL, PostgreSQL, MongoDB)
- Media players (MP3, MP4, VLC)
- API wrappers (convert REST to SOAP)
- Legacy system integration
- Format converters (JSON, XML, CSV)
- UI component adapters
- Logging adapters (different logging libraries)


## Summary: Adapter Pattern

### Structure:

```
┌─────────────────────┐
│  Target Interface   │  ← What client expects
│  + request()        │
└─────────────────────┘
         ▲
         │ implements
         │
┌────────┴────────┐
│    Adapter       │  ← Converts interface
│  - adaptee       │
│  + request()     │
└────────┬─────────┘
         │ uses
         ▼
┌─────────────────────┐
│   Adaptee           │  ← Existing class
│  + specificRequest()│  ← Incompatible interface
└─────────────────────┘
```

### Key Components:

1. **Target Interface**: Interface that client expects
2. **Adaptee**: Existing class with incompatible interface
3. **Adapter**: Class that converts adaptee's interface to target interface
4. **Client**: Uses target interface, doesn't know about adaptee

### Advantages:

✅ **Compatibility**: Makes incompatible interfaces work together  
✅ **Reusability**: Reuse existing classes without modification  
✅ **Single Responsibility**: Adapter has one clear purpose  
✅ **Open/Closed Principle**: Easy to add new adapters  
✅ **Decoupling**: Client doesn't depend on adaptee classes  
✅ **Integration**: Easy to integrate third-party libraries  

### Disadvantages:

❌ **Additional Complexity**: Adds another layer of abstraction  
❌ **More Classes**: Need to create adapter classes  
❌ **Performance Overhead**: Slight overhead due to indirection  

### Remember:

> **"Adapter Pattern allows objects with incompatible interfaces to work together by wrapping an object with an adapter that converts its interface."**

### Adapter vs Other Patterns:

- **Adapter vs Facade**: Adapter converts interface, Facade simplifies interface
- **Adapter vs Decorator**: Adapter changes interface, Decorator adds behavior
- **Adapter vs Proxy**: Adapter converts interface, Proxy controls access

### Python Best Practices:

- Use composition (Object Adapter) over inheritance
- Keep adapters simple - just convert interface
- Use abstract base classes for target interface
- Consider using `@functools.wraps` for function adapters
- Document what interface is being adapted
