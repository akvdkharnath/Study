# Git Reset: Use Cases and Visual Examples

This notebook explains how `git reset` works, when to use it, and provides visual examples for different scenarios involving local and remote branches. It also compares `git reset` to `git revert` and covers advanced options and flags.



## What is `git reset`?

- `git reset` moves the current branch tip (HEAD) to a specified commit.
- It can also change the state of your staging area (index) and working directory, depending on the mode used (`--soft`, `--mixed`, `--hard`).
- Unlike `git revert`, `git reset` can remove commits from history (rewriting history).
- **Warning:** Use with caution, especially on shared branches.



## Common Use Cases for `git reset`

### 1. Undo the Latest Commit (local only)
```bash
git reset --hard HEAD~1
```
- Moves HEAD and branch pointer back by one commit.
- **Removes** the latest commit from history (local only).
- The code, staging area, and working directory all match the new HEAD.

### 2. Undo Multiple Commits (local only)
```bash
git reset --hard HEAD~3
```
- Removes the last 3 commits from history.
- The code, staging area, and working directory all match the new HEAD.

### 3. Reset to a Specific Commit by Commit ID
```bash
git reset --hard <commit-hash>
```
- Moves HEAD and branch pointer to the specified commit.
- All commits after that are removed from history (local only).

### 4. Reset Staging Area Only (keep code changes)
```bash
git reset --soft HEAD~1
```
- Moves HEAD back by one commit, but keeps changes staged.

### 5. Reset Staging Area and Working Directory (keep code changes as unstaged)
```bash
git reset --mixed HEAD~1
```
- Moves HEAD back by one commit, keeps changes in working directory but unstaged.

### 6. Discard All Local Changes and Commits, Match Remote
```bash
git fetch origin
git reset --hard origin/main
```
- Resets your branch to match the remote, discarding all local commits and changes.



## Visual Examples: Local and Remote Branch Status

Below are visual diagrams showing the state of your local and remote branches **before and after** running `git reset` in different scenarios.



### 1. Undo the Latest Commit (local only)

**Before:**
```
Remote:   A -- B -- C  (origin/main)
Local:    A -- B -- C  (main, HEAD)
```

**After `git reset --hard HEAD~1`:**
```
Remote:   A -- B -- C  (origin/main)
Local:    A -- B  (main, HEAD)
```
- Commit C is removed from local history and code.
- **After reset, your real code looks like it did after B.**
- The remote branch is unchanged until you force push (not recommended on shared branches).



### 2. Undo Multiple Commits (local only)

**Before:**
```
Remote:   A -- B -- C  (origin/main)
Local:    A -- B -- C -- D -- E  (main, HEAD)
```

**After `git reset --hard HEAD~3`:**
```
Remote:   A -- B -- C  (origin/main)
Local:    A -- B  (main, HEAD)
```
- Commits C, D, and E are removed from local history and code.
- **After reset, your real code looks like it did after B.**
- The remote branch is unchanged until you force push.



### 3. Reset to a Specific Commit by Commit ID

**Before:**
```
Remote:   A -- B -- C -- D -- E  (origin/main)
Local:    A -- B -- C -- D -- E -- F -- G  (main, HEAD)
```

**After `git reset --hard <commit-hash-of-C>`:**
```
Remote:   A -- B -- C -- D -- E  (origin/main)
Local:    A -- B -- C  (main, HEAD)
```
- Commits after C (D, E, F, G) are removed from local history and code.
- **After reset, your real code looks like it did after C.**
- The remote branch is unchanged until you force push.



### 4. Reset Staging Area Only (keep code changes)

**Before:**
```
A -- B -- C (HEAD)
```
- You have changes staged for commit C.

**After `git reset --soft HEAD~1`:**
```
A -- B (HEAD)
```
- Commit C is removed, but its changes are kept in the staging area (ready to be recommitted).
- **After reset, your real code looks like it did after C, but C is not in history and changes are staged.**



### 5. Reset Staging Area and Working Directory (keep code changes as unstaged)

**Before:**
```
A -- B -- C (HEAD)
```
- You have changes staged for commit C.

**After `git reset --mixed HEAD~1`:**
```
A -- B (HEAD)
```
- Commit C is removed, but its changes are kept in the working directory (unstaged).
- **After reset, your real code looks like it did after C, but C is not in history and changes are unstaged.**



### 6. Discard All Local Changes and Commits, Match Remote

**Before:**
```
Remote:   A -- B -- C  (origin/main)
Local:    A -- B -- C -- D -- E  (main, HEAD)
```
- D and E are local commits, not pushed.

**After `git fetch origin` and `git reset --hard origin/main`:**
```
Remote:   A -- B -- C  (origin/main)
Local:    A -- B -- C  (main, HEAD)
```
- All local commits and changes are discarded.
- **After reset, your real code and history exactly match the remote.**
