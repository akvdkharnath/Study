```
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt


from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```
df = pd.read_csv("https://raw.githubusercontent.com/akvdkharnath/My-Research/master/Python%20Files/Data/Salary_Data.csv")
df
```

```
x = df.iloc[:,:-1]
y = df.iloc[:,-1]
print(x)
print(y)
```

```
x_train,x_test,y_train,y_test  = train_test_split(x,y,test_size = 0.8)
```

```
lr = LinearRegression()
lr.fit(x_train,y_train)
y_predict = lr.predict(x_test)
```

```
plt.scatter(x_train, y_train, color = 'red')
print(plt.plot(x_train, lr.predict(x_train), color = 'blue'))


plt.scatter(x_test, y_test, color = 'red')
print(plt.plot(x_test, lr.predict(x_test), color = 'blue'))

print(lr.predict([[12]]))
print(lr.coef_)
print(lr.intercept_)
```
