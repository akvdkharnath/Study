```
import pandas as pd
import numpy as np 

from sklearn.preprocessing import OneHotEncoder,LabelEncoder,MinMaxScaler,StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
```

Filling empty values with mean of column 

```
df = pd.read_csv("https://raw.githubusercontent.com/akvdkharnath/My-Research/master/Python%20Files/Data/Data.csv")
df = df.fillna(df.mean())
df
```

split input and output data

```
x = df.iloc[:,:-1]
y = df.iloc[:,-1]
print(x)
print(y)
```

**One Hot Encodeing:**

used to encode cetogorical values in to many columns 
ColumnTransformer -->

transformers --> [(encoding method,function,[columns to apply])]

remainder --> if passthrough it will result onehotencoded columns + other columns, if drop then only return onehotencoded columns 



```
ct = ColumnTransformer(transformers=[("encoder",OneHotEncoder(),[0])],remainder="passthrough")
x = np.array(ct.fit_transform(x),dtype=str)
x = pd.DataFrame(x)
print(x)
```

**Label Encoding:**

used to encode a cetogorical values into numarical values 

difference between OneHot Encoding and Label Encoding
OneHot is used to encode cetogorical column into mutiple columns but Label encoder is used to encode cetogorical column into singlr column  

```
le = LabelEncoder()
y = np.array(le.fit_transform(y))
y
```

**Split data into Train and Test**

its better to have data in 8:2 ratio

```
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.8)
```

**Scaling:**

1. Normalization
    1. converts data into 0 to 1 
    2. use MinMaxscaler to apply
    3. apply only if data is in Normal/ Gaussion distribution

2. Standradization
    1. converts data from -3 to +3 
    2. use StandardScaler to apply 
    3. can apply for all most all data

**Note:** pls dont apply scaling for One hot encoded data 


```
sc = StandardScaler()
x_train.iloc[:,3:] = sc.fit_transform(x_train.iloc[:,3:])
x_test
```
